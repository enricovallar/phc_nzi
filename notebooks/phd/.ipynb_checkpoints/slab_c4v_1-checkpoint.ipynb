```python

# --- Arrow colors ---
COLOR_RE = 'black'
COLOR_IM = '#00E676'        # bright green
EDGE_RE  = 'white'
EDGE_IM  = 'black'

# # Build a parameter-stamped subfolder: figures/r1_XXXX_r2_XXXX_eps_XX
# param_tag = f"r1_{r1:.4f}_r2_{r2:.4f}_eps_{eps_inp:.2f}"
# SAVE_DIR = os.path.join(os.path.dirname(os.path.abspath(".")), "phd", "figures", param_tag)
# os.makedirs(SAVE_DIR, exist_ok=True)
# print(f"Saving figures to: {SAVE_DIR}")


class ArrowHandler(HandlerPatch):
    """Custom legend handler that draws an arrow instead of a rectangle."""
    def create_artists(self, legend, orig_handle, xdescent, ydescent,
                       width, height, fontsize, trans):
        arrow = FancyArrow(
            xdescent, ydescent + height / 2,
            width * 0.9, 0,
            width=height * 0.35,
            head_width=height * 0.9,
            head_length=width * 0.25,
            fc=orig_handle.get_facecolor(),
            ec=orig_handle.get_edgecolor(),
            linewidth=0.8,
            transform=trans
        )
        return [arrow]


def add_arrow_legend(fig_or_ax, is_figure=False, **kwargs):
    """Add arrow-style legend for Re/Im E_parallel."""
    arrow_re = FancyArrow(0, 0.5, 0.5, 0, width=0.08,
                          head_width=0.25, head_length=0.12,
                          fc=COLOR_RE, ec=EDGE_RE)
    arrow_im = FancyArrow(0, 0.5, 0.5, 0, width=0.08,
                          head_width=0.25, head_length=0.12,
                          fc=COLOR_IM, ec=EDGE_IM)
    target = fig_or_ax.legend if is_figure else fig_or_ax.legend
    return target(
        [arrow_re, arrow_im],
        [r'$\mathrm{Re}(E_{\parallel})$', r'$\mathrm{Im}(E_{\parallel})$'],
        handler_map={FancyArrow: ArrowHandler()},
        ncol=2, 
        framealpha=0.95, facecolor='white', edgecolor='black',
        **kwargs
    )


def plot_field_quiver_on_ax(sim: Simulation, conv_opt: MPBDataOptions = MPBDataOptions(rectify=True, periods=1, transpose=False),
                            mode: str = "te", k_idx: int = 12, b_idx: int = 0, ax=None, step = 2):
    """Plot H_z background with E-field quiver arrows on a given axes."""

    if ax is None: 
        ax = plt.gca()
    
    if mode.lower() == "te":
        field_z = sim.load_and_convert_field_data(
            k_idx, b_idx, "z", mode, "h", conversion_options=conv_opt, overwrite=True)
        field_x = sim.load_and_convert_field_data(
            k_idx, b_idx, "x", mode, "e", conversion_options=conv_opt, overwrite=True)
        field_y = sim.load_and_convert_field_data(
            k_idx, b_idx, "y", mode, "e", conversion_options=conv_opt, overwrite=True)
    elif mode.lower() == "tm":
        field_z = sim.load_and_convert_field_data(
            k_idx, b_idx, "z", mode, "e", conversion_options=conv_opt, overwrite=True)
        field_x = sim.load_and_convert_field_data(
            k_idx, b_idx, "x", mode, "h", conversion_options=conv_opt, overwrite=True)
        field_y = sim.load_and_convert_field_data(
            k_idx, b_idx, "y", mode, "h", conversion_options=conv_opt, overwrite=True)
    else:
        raise ValueError(f"Unsupported mode: {mode}. Use 'te' or 'tm'.")

    field_z_real = np.real(field_z)
    vmax_h = np.max(np.abs(field_z_real))
    im = ax.imshow(field_z_real, interpolation='spline36', cmap="RdBu_r",
                   vmin=-vmax_h, vmax=vmax_h, origin='lower')


    ny, nx = field_z.shape
    x = np.arange(nx)
    y = np.arange(ny)
    X, Y = np.meshgrid(x, y)
    X_sub = X[::step, ::step]
    Y_sub = Y[::step, ::step]

    field_x_r = np.real(field_x)[::step, ::step]
    field_y_r = np.real(field_y)[::step, ::step]
    field_x_i = np.imag(field_x)[::step, ::step]
    field_y_i = np.imag(field_y)[::step, ::step]

    mag_real = np.sqrt(field_x_r**2 + field_y_r**2)
    mag_imag = np.sqrt(field_x_i**2 + field_y_i**2)
    max_mag = max(mag_real.max(), mag_imag.max())
    arrow_scale = max_mag * 15

    ax.quiver(X_sub, Y_sub, field_x_r, field_y_r,
              color=COLOR_RE, edgecolor=EDGE_RE, linewidth=0.3,
              scale=arrow_scale, width=0.004, headwidth=4, headlength=4)

    offset = 0.3
    ax.quiver(X_sub + offset, Y_sub + offset, field_x_i, field_y_i,
              color=COLOR_IM, edgecolor=EDGE_IM, linewidth=0.3,
              scale=arrow_scale, width=0.004, headwidth=4, headlength=4)

    try:
        eps_data = sim.load_and_convert_epsilon_data(options=conv_opt)
        ax.contour(eps_data, levels=[eps_data.max() * 0.5],
                   colors='gray', linewidths=1.5, linestyles='--',
                   alpha=0.7, origin='lower')
    except Exception:
        pass

    ax.set_xlabel("x (grid pts)")
    ax.set_ylabel("y (grid pts)")
    ax.set_xlim(0, nx)
    ax.set_ylim(0, ny)
    ax.set_aspect('equal')
    return im


```

```python
# Run Bayesian Optimization
lsf_config = LSFJobConfiguration(num_processors=numproc, span_option="hosts", span_value=1, queue="hpc",   mem = "1GB")

from phc_nzi.mpi_bayesian_optimization import MPIBayesianOptimizator
numproc = 24


bo_options = {
    "dimensions": [(0.2, 0.4), (0.2, 0.4)],
    "acq_func": "LCB",
    "acq_func_kwargs": {"kappa": 2.5},
    "n_initial_points": numproc * 20,
    "initial_point_generator": "lhs"
}

optimizer = MPIBayesianOptimizator(
        simulation_name=simulation_2D_opt_456.simulation_name,
        scheme_script=simulation_2D_opt_456.script, 
        directory=simulation_2D_opt_456.directory,
        maxiter=30, 
        batch_size=numproc,
        param_names=["r1", "r2"],
        polarization="te",
        bands=[4,5,6],
        bo_options=bo_options, # <-- The JSON magic happens here
    )

if SIMULATE is True:
    optimizer.erease_data_file()
    optimizer.submit_lsf_job(lsf_config, conda_env_name="mpb-nzi-env")
```

```python
# Plot the optimization data and the GP surrogate predictions in the r1-r2 plane
# 1. Load the Analyzer and the Saved Optimizer
analyzer = OptimizationDataAnalyzer(optimizer.data_file)
analyzer.load_data()
loaded_optimizer = joblib.load(optimizer.model_file)

# Collect raw data and compute filling factor
df = analyzer.get_points_above_treshold(0)
df["filling_factor"] = (np.pi * df["r1"]**2 + np.pi * df["r2"]**2) / 1**2

# Prepare predictions from the GP
gp_model = loaded_optimizer.models[-1]
valid = (analyzer.cost_vals > 0) & (~np.isnan(analyzer.cost_vals))
X_train = np.c_[analyzer.param1_vals[valid], analyzer.param2_vals[valid]]

r1_grid = np.linspace(0.2, 0.4, 200)
r2_grid = np.linspace(0.2, 0.4, 200)
R1, R2 = np.meshgrid(r1_grid, r2_grid)
grid_points = np.c_[R1.ravel(), R2.ravel()]

grid_points_transformed = loaded_optimizer.space.transform(grid_points)
predicted_cost = gp_model.predict(grid_points_transformed)

# Diagnostics: inspect predicted cost distribution (helps explain extreme FOM values)
pc = predicted_cost
try:
    pc_min = np.nanmin(pc)
    pc_p1 = np.nanpercentile(pc, 1)
    pc_med = np.nanmedian(pc)
    pc_p99 = np.nanpercentile(pc, 99)
    pc_max = np.nanmax(pc)
    print(f"predicted_cost: min={pc_min:.3e}, 1%={pc_p1:.3e}, med={pc_med:.3e}, 99%={pc_p99:.3e}, max={pc_max:.3e}")
    n_nonpos = np.sum(np.asarray(pc) <= 0)
    print('predicted_cost non-positive count:', int(n_nonpos))
except Exception as e:
    print('Failed to compute predicted_cost diagnostics:', e)

# Calculate the Figure of Merit (safely handle non-positive predicted_cost)
fom_raw = (1.0 / df["cost"])
predicted_cost_safe = np.where(predicted_cost > 0, predicted_cost, np.nan)
Predicted_FOM_2D = (1.0 / predicted_cost_safe).reshape(R1.shape)

# 2. Extract Global Min/Max and setup LogNorm
# sanitize FOM values (remove non-positive / inf / nan)
all_fom_vals = np.r_[fom_raw.values, Predicted_FOM_2D.ravel()]
finite_pos = all_fom_vals[np.isfinite(all_fom_vals) & (all_fom_vals > 0)]
if finite_pos.size == 0:
    raise RuntimeError("No positive finite FOM values to plot.")
# use robust vmin/vmax to avoid outliers (1st/99th percentiles)
vmin = np.percentile(finite_pos, 1)
vmax = np.percentile(finite_pos, 99)
vmin = max(vmin, 1e-12)
log_norm = mcolors.LogNorm(vmin=vmin, vmax=vmax)

# Initialize figure and axes
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=ROW1_COL2_SIZE)

# INCREASE SEPARATION BETWEEN PLOTS
plt.subplots_adjust(wspace=0.35)

# ---------------------------------------------------------
# Plot 1: Raw Data (Left)
# ---------------------------------------------------------
# Pass the log_norm to the scatter plot
scatter_raw = ax1.scatter(df["r1"], df["r2"], c=fom_raw, cmap="cool", norm=log_norm, s=2, marker='o')

ff_min, ff_max = df["filling_factor"].min(), df["filling_factor"].max()
threshold_ff = (ff_min + ff_max) / 2

best_point_below_threshold = df.loc[df[df["filling_factor"] < threshold_ff]["cost"].idxmin()]
best_point_above_threshold = df.loc[df[df["filling_factor"] >= threshold_ff]["cost"].idxmin()]

ax1.scatter(best_point_below_threshold["r1"], best_point_below_threshold["r2"], color="k", label=rf"Best $FF<{threshold_ff:0.2f}$", marker="o", s=30)
ax1.scatter(best_point_above_threshold["r1"], best_point_above_threshold["r2"], color="k", label=rf"Best $FF\geq{threshold_ff:0.2f}$", marker="*", s=30)

ax1.set_xlabel("$r_1/a$")
ax1.set_ylabel("$r_2/a$")
ax1.set_title("Optimization data")
ax1.grid(alpha=0.4)
ax1.set_aspect('equal')

# ---------------------------------------------------------
# Plot 2: GP Surrogate (Right)
# ---------------------------------------------------------
# Use contourf with the LogNorm (levels chosen in log-space)
levels = np.logspace(np.log10(vmin), np.log10(vmax), 200)
heatmap = ax2.contourf(R1, R2, Predicted_FOM_2D, levels=levels, cmap="cool", norm=log_norm, extend='both')

ridge_threshold = np.percentile(Predicted_FOM_2D, 60)
ridge_mask = Predicted_FOM_2D > ridge_threshold
r1_ridge = R1[ridge_mask]
r2_ridge = R2[ridge_mask]
weights_ridge = Predicted_FOM_2D[ridge_mask]

conic_params = analyzer.fit_ellipse(r1_ridge, r2_ridge, w=weights_ridge)
analyzer.plot_ellipse_from_conic(*conic_params, ax=ax2, plot_kwds={'color': 'limegreen', 'label': 'Degeneracy Ellipse'})

ax2.scatter(X_train[:, 0], X_train[:, 1], c='white', edgecolors='black', s=5, alpha=0.5, label="Sampled Points")

ax2.set_xlabel("$r_1/a$")
ax2.set_ylabel("$r_2/a$")
ax2.set_title("BO Surrogate Map")
ax2.grid(alpha=0.4)
ax2.set_aspect('equal')

# ---------------------------------------------------------
# Global Adjustments (Ticks, Colorbar, Legend)
# ---------------------------------------------------------


for ax in (ax1, ax2):
    ax.xaxis.set_major_formatter(FormatStrFormatter('%.2f'))
    ax.yaxis.set_major_formatter(FormatStrFormatter('%.2f'))


# Create a shared colorbar from the contourf mappable so triangular end-caps appear on both subplots
cbar = fig.colorbar(heatmap, ax=[ax1, ax2], fraction=0.035, pad=0.04, extend='both', format=LogFormatterSciNotation())
cbar.set_label(r"1/$\mathrm{C}$")
cbar.ax.tick_params(labelsize=9)
cbar.locator = LogLocator(base=10.0, numticks=6)
cbar.update_ticks()
# Add minor log ticks to give a glimpse of progress between decades
cbar.ax.yaxis.set_minor_locator(LogLocator(base=10.0, subs=np.arange(2, 10), numticks=12))
cbar.ax.yaxis.set_minor_formatter(plt.NullFormatter())
cbar.ax.tick_params(which='minor', length=4)

handles1, labels1 = ax1.get_legend_handles_labels()
handles2, labels2 = ax2.get_legend_handles_labels()

# Legend placement (pushed slightly down to avoid overlapping the x-axis)
fig.legend(handles1 + handles2, labels1 + labels2, loc='lower center', bbox_to_anchor=(0.5, -0.15), ncol=4, framealpha=0.95, facecolor='white', edgecolor='black')
plt.savefig(os.path.join(figure_path, "optimization_data_bo_surrogate_2D.png"))
plt.show()
#display description of the figure
description = """
Figure: Left panel shows the raw optimization data points in the r1-r2 parameter space, colored by their Figure of Merit (FOM = 1/cost) on a logarithmic scale. Red markers indicate the best points below and above the filling factor threshold. Right panel shows the Gaussian Process surrogate model's predicted FOM landscape across the parameter space, with a fitted ellipse highlighting the degeneracy ridge. Sampled points are overlaid as white circles with black edges. A shared colorbar indicates the FOM values on a log scale.
"""
print(description)
```

### Band Linearity Analysis
Evaluate the linearity of the top Dirac-cone band (band 6) for points along the optimization ellipse.

```python
## Find the points along the degeneracy ellipse that are closest to Gamma, and run simulations there to confirm the Dirac-like dispersion
# Create a lean configuration: only frequencies, no field output
config_linearity = dict(
    resolution=RESOLUTION,
    num_bands=8,
    k_points=k_points_2D_optimized["k_points_values"],
    k_points_interpolation_factor=60,
)
mpb_config_lin = MPBSchemeConfigurator(photonic_crystal_2D, ["te"], **config_linearity)
script_lin = mpb_config_lin.get_scheme_config(join_newline=True)

# Sample points along the Dirac-like ellipse from optimization data
threshold_lin = 500
n_samples = 100
r1_ell, r2_ell, f_ell, conic_ell = analyzer.compute_freq_dirac_along_ellipse(
    threshold=threshold_lin, n_points=n_samples)

# --- Restrict to the symmetric half: r1 <= r2 (r1 and r2 are interchangeable) ---
mask_sym = r1_ell <= r2_ell
r1_ell = r1_ell[mask_sym]
r2_ell = r2_ell[mask_sym]
f_ell = f_ell[mask_sym]

print(f"Sampling {len(r1_ell)} points on the half-ellipse (r1 ≤ r2)")
print(f"  r1 range: [{r1_ell.min():.4f}, {r1_ell.max():.4f}]")
print(f"  r2 range: [{r2_ell.min():.4f}, {r2_ell.max():.4f}]")

# --- BO Surrogate Map + Half-Ellipse Plot ---
fig, ax = plt.subplots(1, 1, figsize=ROW1_COL1_SIZE)

# Contourf of the GP surrogate (reusing variables from the earlier BO surrogate cell)
levels_lin = np.logspace(np.log10(vmin), np.log10(vmax), 200)
heatmap_lin = ax.contourf(R1, R2, Predicted_FOM_2D, levels=levels_lin,
                          cmap="cool", norm=log_norm, extend='both')
cbar_lin = fig.colorbar(heatmap_lin, ax=ax, fraction=0.035, pad=0.04,
                        extend='both', format=LogFormatterSciNotation())
cbar_lin.set_label(r"1/$\mathrm{C}$")
cbar_lin.ax.tick_params(labelsize=9)
cbar_lin.locator = LogLocator(base=10.0, numticks=6)
cbar_lin.update_ticks()
cbar_lin.ax.yaxis.set_minor_locator(LogLocator(base=10.0, subs=np.arange(2, 10), numticks=12))
cbar_lin.ax.yaxis.set_minor_formatter(plt.NullFormatter())
cbar_lin.ax.tick_params(which='minor', length=4)

# Compute the full ellipse parametrically from conic coefficients
A_e, B_e, C_e, D_e, E_e, F_e = conic_ell
denom_e = B_e**2 - 4 * A_e * C_e
x0_e = (2 * C_e * D_e - B_e * E_e) / denom_e
y0_e = (2 * A_e * E_e - B_e * D_e) / denom_e
theta_e = 0.5 * np.arctan2(B_e, A_e - C_e)
num_e = 2 * (A_e * x0_e**2 + B_e * x0_e * y0_e + C_e * y0_e**2 - F_e)
term_e = np.sqrt((A_e - C_e)**2 + B_e**2)
ae = np.sqrt(num_e / (A_e + C_e - term_e))
be = np.sqrt(num_e / (A_e + C_e + term_e))
if be > ae:
    ae, be = be, ae

t_full = np.linspace(0, 2 * np.pi, 500)
ell_x = x0_e + ae * np.cos(t_full) * np.cos(theta_e) - be * np.sin(t_full) * np.sin(theta_e)
ell_y = y0_e + ae * np.cos(t_full) * np.sin(theta_e) + be * np.sin(t_full) * np.cos(theta_e)

# Only plot the half-ellipse where r1 <= r2 (handle possible wrap-around)
half_mask = ell_x <= ell_y
idx_half = np.where(half_mask)[0]
if len(idx_half) > 0:
    breaks = np.where(np.diff(idx_half) > 1)[0] + 1
    segments = np.split(idx_half, breaks)
    for k, seg in enumerate(segments):
        ax.plot(ell_x[seg], ell_y[seg], color='limegreen', linewidth=2,
                label='Degeneracy half-ellipse' if k == 0 else None)

# Symmetry line r1 = r2
r_lim = np.array([min(r1_grid.min(), r2_grid.min()), max(r1_grid.max(), r2_grid.max())])
ax.plot(r_lim, r_lim, 'w--', linewidth=1, alpha=0.7, label='$r_1 = r_2$')

# Sampled points on the half-ellipse
ax.plot(r1_ell, r2_ell, color = "darkred", linestyle = "", marker = "o", label="Sampled points", markersize=2)

ax.set_xlabel("$r_1/a$")
ax.set_ylabel("$r_2/a$")
ax.set_title("BO Surrogate + Degeneracy Half-Ellipse, " + rf"$\varepsilon={eps_inp}$")
ax.set_aspect('equal')
ax.grid(alpha=0.4)
ax.xaxis.set_major_formatter(FormatStrFormatter('%.2f'))
ax.yaxis.set_major_formatter(FormatStrFormatter('%.2f'))
ax.legend(loc='lower center', bbox_to_anchor=(0.5, -0.35), ncol=2,
          framealpha=0.95, facecolor='white', edgecolor='black')
plt.show()

# Create a unique Simulation object for each half-ellipse point
sims_lin = []
for i, (r1_i, r2_i) in enumerate(zip(r1_ell, r2_ell)):
    sim_name = f"{name_2D_optimized}_lin_{i:03d}"
    sim = Simulation(
        simulation_name=sim_name,
        script=script_lin,
        directory=os.path.join(data_rooth, sim_name)
    )
    sims_lin.append({'r1': r1_i, 'r2': r2_i, 'sim': sim, 'f_dirac_approx': f_ell[i]})

# Run all simulations concurrently
if SIMULATE is True:
    from concurrent.futures import ThreadPoolExecutor, as_completed
    N_WORKERS = 8

    def run_lin_entry(args):
        j, entry = args
        entry['sim'].run_hpc(
            mpb_command_line_params=dict(r1=entry['r1'], r2=entry['r2']),
            mpi=True, load_epsilon=False)
        return j

    print(f"Running {len(sims_lin)} simulations with {N_WORKERS} workers...")
    with ThreadPoolExecutor(max_workers=N_WORKERS) as pool:
        futures = {pool.submit(run_lin_entry, (j, e)): j for j, e in enumerate(sims_lin)}
        for fut in as_completed(futures):
            j = fut.result()
            print(f"  [{j+1}/{len(sims_lin)}] r1={sims_lin[j]['r1']:.4f}, r2={sims_lin[j]['r2']:.4f} done")
    print(f"\nAll {len(sims_lin)} simulations completed.")
else:
    print(f"SIMULATE=False → Skipping actual runs. Using existing data if available, otherwise will error out when trying to load frequencies.")
```

```python
## Make sure the points are degenerate at Gamma

# ================================================================
# For each of the 40 ellipse points, fix r1 and sweep r2 at Γ
# to find the EXACT band 4–6 degeneracy — using parallel ParamSweepers
# ================================================================
import time
from concurrent.futures import ThreadPoolExecutor, as_completed
from phc_nzi.sweeper import ParamSweeper

n_r2_sweep = 200         # number of r2 values per point
r2_half_width = 0.01    # search range: r2_original ± this
N_WORKERS = 25            # number of sweepers to run in parallel

# Gamma-only configuration → very fast per simulation
config_deg = dict(
    resolution=RESOLUTION,
    num_bands=8,
    k_points=[mp.Vector3(0, 0, 0)],
)
mpb_config_deg = MPBSchemeConfigurator(photonic_crystal_2D, ["te"], **config_deg)
script_deg = mpb_config_deg.get_scheme_config(join_newline=True)

total_sims = len(r1_ell) * n_r2_sweep
print(f"Sweeping r2 for {len(r1_ell)} points x {n_r2_sweep} r2 values = {total_sims} Γ-only simulations")

# Build one ParamSweeper per ellipse point
sweeper_entries = []
for i in range(len(r1_ell)):
    r1_i = r1_ell[i]
    r2_orig = r2_ell[i]
    r2_lo = max(0.05, r2_orig - r2_half_width)
    r2_hi = min(0.49, r2_orig + r2_half_width)
    r2_sweep = np.linspace(r2_lo, r2_hi, n_r2_sweep)

    sim_deg = Simulation(
        simulation_name=f"{name_2D_optimized}_deg_sweep_{i:03d}",
        script=script_deg,
        directory=os.path.join(data_rooth, f"{name_2D_optimized}_deg_sweep_{i:03d}")
    )
    sweeper = ParamSweeper(
        simulation=sim_deg,
        param_name="r2",
        values=r2_sweep.tolist(),
        polarization="te",
        bands=[4, 6],
        other_command_line_params={"r1": r1_i}
    )
    sweeper_entries.append({
        'idx': i, 'r1': r1_i, 'r2_orig': r2_orig,
        'r2_sweep': r2_sweep, 'sweeper': sweeper
    })

t0 = time.time()

def run_and_analyze(entry):
    """Run one sweeper and return the corrected-point dict."""
    sw = entry['sweeper']
    if SIMULATE:
        sw.run()
    data = sw.data if len(sw.data) > 0 else sw.load_df()
    gaps = (data['band_6'] - data['band_4']).abs().values
    best = int(np.argmin(gaps))
    r2_corr = entry['r2_sweep'][best]

    if best == 0 or best == n_r2_sweep - 1:
        print(f"  ⚠ [{entry['idx']+1:2d}/{len(r1_ell)}] r1={entry['r1']:.4f}: "
              f"minimum at EDGE of sweep range — consider widening r2_half_width")

    return {
        'idx': entry['idx'],
        'r1': entry['r1'],
        'r2_original': entry['r2_orig'],
        'r2_corrected': r2_corr,
        'gap_corrected': gaps[best],
        'delta_r2': r2_corr - entry['r2_orig'],
    }

if SIMULATE:
    print(f"Running {N_WORKERS} sweepers in parallel...")
    corrected_points = []
    with ThreadPoolExecutor(max_workers=N_WORKERS) as pool:
        futures = {pool.submit(run_and_analyze, e): e for e in sweeper_entries}
        for fut in as_completed(futures):
            result = fut.result()
            corrected_points.append(result)
            idx = futures[fut]['idx']
            elapsed = time.time() - t0
            print(f"  [{idx+1:2d}/{len(r1_ell)}] r1={result['r1']:.4f}: "
                  f"r2={result['r2_original']:.4f} → {result['r2_corrected']:.4f} ")
    corrected_points.sort(key=lambda x: x['idx'])
else:
    print(f"SIMULATE=False → Loading existing sweeper data with {N_WORKERS} workers...")
    corrected_points = []
    with ThreadPoolExecutor(max_workers=N_WORKERS) as pool:
        futures = {pool.submit(run_and_analyze, e): e for e in sweeper_entries}
        for fut in as_completed(futures):
            result = fut.result()
            corrected_points.append(result)
            print(f"  [{result['idx']+1:2d}/{len(r1_ell)}] r1={result['r1']:.4f}: "
                  f"r2={result['r2_original']:.4f} → {result['r2_corrected']:.4f} "
                  f"(Δ={result['delta_r2']:+.5f}, gap={result['gap_corrected']:.2e})")
    corrected_points.sort(key=lambda x: x['idx'])

df_corr = pd.DataFrame(corrected_points)
print(f"\nDone in {time.time()-t0:.0f}s")
print(f"Mean |Δr2| = {df_corr['delta_r2'].abs().mean():.5f}")
print(f"Max  |Δr2| = {df_corr['delta_r2'].abs().max():.5f}")
print(f"Mean gap_corrected = {df_corr['gap_corrected'].mean():.2e}")

# ================================================================
# Plot: Original vs Corrected ellipse
# ================================================================
fig, ax = plt.subplots(1, 1, figsize=ROW1_COL1_SIZE)



# Approximated half-ellipse and corrected points (markers only, no closing line)
ax.plot(r1_ell, r2_ell, "ro", ms=2, label="Approximated", alpha=0.7)
# Filter corrected points to only valid ones (same criterion as cell 27)
valid_mask = (df_corr['r1'] + df_corr['r2_corrected']) < (np.sqrt(2) / 2)
valid_mask &= ~np.isclose(df_corr['r1'] + df_corr['r2_corrected'], np.sqrt(2) / 2, atol=0.01)
df_corr_valid = df_corr[valid_mask]
df_corr_invalid = df_corr[~valid_mask]

ax.plot(df_corr_valid['r1'].values, df_corr_valid['r2_corrected'].values, "go", ms=2,
        label=f"Corrected")
if len(df_corr_invalid) > 0:
    ax.plot(df_corr_invalid['r1'].values, df_corr_invalid['r2_corrected'].values, 
            "o", color="gray", ms=2, label=f"Invalid", alpha = 0.7)
ax.set_xlabel("$r_1/a$")
ax.set_ylabel("$r_2/a$")
ax.set_title(f"Sample Points \n 2D Photonic Crystal Slab, $\\varepsilon={eps_inp}$")
ax.set_aspect("equal")
ax.grid(alpha=0.4)
ax.xaxis.set_major_formatter(FormatStrFormatter('%.2f'))
ax.yaxis.set_major_formatter(FormatStrFormatter('%.2f'))
# Symmetry line r1 = r2
r_lim = np.array([min(r1_grid.min(), r2_grid.min()), max(r1_grid.max(), r2_grid.max())])
ax.plot(r_lim, r_lim, 'k--', linewidth=1, alpha=0.5, label='$r_1 = r_2$')

ax.legend(loc='lower center', bbox_to_anchor=(0.5, -0.40), ncol=3,
          framealpha=0.95, facecolor='white', edgecolor='black')
plt.savefig(os.path.join(figure_path, f"corrected_degeneracy_points.png"))
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import os
from scipy.interpolate import interp1d

b_idxs = [4, 5, 6]
target = len(sims_lin) // 2

for i, entry in enumerate(sims_lin):
    df_freq = entry['sim'].load_frequency_data("te")
    gamma_idx = df_freq[(df_freq['k1'] == 0) & (df_freq['k2'] == 0)].index[0]
    entry['gamma_idx'] = gamma_idx
    entry['df_freq'] = df_freq
    entry['f_dirac'] = df_freq["te band 5"].values[gamma_idx]
    entry['k_vecs'] = [(df_freq['k1'].values[j], df_freq['k2'].values[j]) for j in range(len(df_freq))]
    entry['k_idxs'] = list(range(len(df_freq)))
    entry['bands'] = {f"band_{b}": df_freq[f"te band {b}"].values for b in b_idxs}
    entry['ff'] = 1 - (np.pi * entry['r1']**2 + np.pi * entry['r2']**2)
    entry['is_valid'] = (entry['r1'] + entry['r2']) < (np.sqrt(2) / 2) and not np.isclose(entry['r1'] + entry['r2'], np.sqrt(2) / 2, atol=0.01)

    # --- Gamma -> X Direction (ky == 0) ---
    ky_zero_idxs = [j for j, (k1, k2) in enumerate(entry['k_vecs']) if abs(k2) == 0]
    kx_vals = np.array([entry['k_vecs'][j][0] for j in ky_zero_idxs])
    freqs_band4_X = np.array([entry['bands']['band_4'][j] for j in ky_zero_idxs])
    freqs_band5_X = np.array([entry['bands']['band_5'][j] for j in ky_zero_idxs])
    freqs_band6_X = np.array([entry['bands']['band_6'][j] for j in ky_zero_idxs])

    coeffs_X = np.polyfit(kx_vals, freqs_band6_X, deg=3)
    poly_X = np.poly1d(coeffs_X)
    slope_6_X = poly_X.deriv()(0) 

    # --- Gamma -> M Direction (|kx| == |ky|) ---
    kM_idxs = [j for j, (k1, k2) in enumerate(entry['k_vecs']) if np.isclose(abs(k1), abs(k2))]
    kM_vals = np.array([np.sign(entry['k_vecs'][j][0]) * np.sqrt(entry['k_vecs'][j][0]**2 + entry['k_vecs'][j][1]**2) for j in kM_idxs])
    freqs_band4_M = np.array([entry['bands']['band_4'][j] for j in kM_idxs])
    freqs_band5_M = np.array([entry['bands']['band_5'][j] for j in kM_idxs])
    freqs_band6_M = np.array([entry['bands']['band_6'][j] for j in kM_idxs])

    coeffs_M = np.polyfit(kM_vals, freqs_band6_M, deg=3)
    poly_M = np.poly1d(coeffs_M)
    slope_6_M = poly_M.deriv()(0) 

    entry["group_velocity_band_6_X"] = slope_6_X
    entry["group_velocity_band_6_M"] = slope_6_M
    
    # --- Analytical Polyfit Isotropy Metric ---
    # 1. Get coefficients (a3, a2, a1, a0)
    a3_X, a2_X, a1_X, a0_X = coeffs_X
    a3_M, a2_M, a1_M, a0_M = coeffs_M
    
    # Define shared evaluation boundary
    k_max = min(np.max(np.abs(kx_vals)), np.max(np.abs(kM_vals)))
    
    # 2. Helper function for the exact polynomial integral area
    def poly_area(a1, a2, a3, k):
        return (a1/2)*k**2 + (a2/3)*k**3 + (a3/4)*k**4
        
    # 3. Calculate exact deviation areas for both directions
    area_dev_X = abs(poly_area(a1_X, a2_X, a3_X, k_max))
    area_dev_M = abs(poly_area(a1_M, a2_M, a3_M, k_max))
    area_dev = 0.5 * (area_dev_X + area_dev_M)
    
    # 4. Calculate exact difference area using the subtracted coefficients
    da1 = a1_X - a1_M
    da2 = a2_X - a2_M
    da3 = a3_X - a3_M
    area_diff = abs(poly_area(da1, da2, da3, k_max))
    
    # 5. Final Metric (1.0 = perfect match, 0.0 = total mismatch)
    entry["isotropy"] = max(0.0, 1.0 - (area_diff / (area_dev + 1e-12)))

    plot_condition = False

        
    if plot_condition:
        plt.figure(figsize=ROW1_COL1_SIZE)  
        plt.plot(kx_vals, freqs_band4_X, "k.-", label="Band 4 (Γ→X)", ms=2)
        plt.plot(kx_vals, freqs_band5_X, "k.-", label="Band 5 (Γ→X)", ms=2)
        plt.plot(kx_vals, freqs_band6_X, "k.-", label="Band 6 (Γ→X)", ms=4)
        plt.plot(kx_vals, poly_X(kx_vals), "r-", label=f"Fit Γ→X ($v_g$={slope_6_X:.4f})")
        
        plt.plot(kM_vals, freqs_band4_M, "gray", marker=".", linestyle="none", label="Band 4 (Γ→M)", ms=2)
        plt.plot(kM_vals, freqs_band5_M, "gray", marker=".", linestyle="none", label="Band 5 (Γ→M)", ms=2)
        plt.plot(kM_vals, freqs_band6_M, "gray", marker=".", linestyle="none", label="Band 6 (Γ→M)", ms=4)
        plt.plot(kM_vals, poly_M(kM_vals), "b-", label=f"Fit Γ→M ($v_g$={slope_6_M:.4f})")

        plt.xlabel("$|k|$ from Γ")
        plt.ylabel("$\omega a/2\pi c$")
        plt.title(f"Dirac Cone Fits for entry {i} (FF {entry['ff']:.4f})\nWhole-Band Isotropy: {entry['isotropy']:.4f}")
        plt.legend()
        plt.grid(True)
        plt.show()

# --- Global Figures ---
valid_entries = [e for e in sims_lin if e['is_valid']]
ff_values = [e['ff'] for e in valid_entries]

# Band Gap Plot
gap_values = [abs(e['bands']['band_6'][e['gamma_idx']] - e['bands']['band_4'][e['gamma_idx']]) for e in valid_entries]
b5_values = [e['bands']['band_5'][e['gamma_idx']] for e in valid_entries]
gap_values_norm = np.array(gap_values) / np.array(b5_values)
# Find the outliers (use mean and std)
gap_mean = np.mean(gap_values_norm)
gap_std = np.std(gap_values_norm)
outlier_mask = np.abs(gap_values_norm - gap_mean) > 0.8 * gap_std
# Mark outliers in the plot
plt.figure(figsize=ROW1_COL1_SIZE)
plt.scatter(np.array(ff_values)[~outlier_mask], gap_values_norm[~outlier_mask], c='purple', edgecolors='k')
plt.scatter(np.array(ff_values)[outlier_mask], gap_values_norm[outlier_mask], c='red', edgecolors='k', marker = 'x')
# Plot invalid as gray dot
plt.xlabel("Filling factor")
plt.ylabel("$\Delta \omega^\Gamma_{6,4}/\omega_5^\Gamma$")
plt.title("Band gap at Γ for corrected points")
plt.yscale('log')
plt.grid(True)
plt.show()


# Group Velocities Plot
plt.figure(figsize=ROW1_COL1_SIZE)
plt.scatter(np.array(ff_values)[~outlier_mask], 
            np.array([e['group_velocity_band_6_X'] for e in valid_entries])[~outlier_mask], 
            c='red', edgecolors='k', label='Γ→X')
plt.scatter(np.array(ff_values)[~outlier_mask], 
            np.array([e['group_velocity_band_6_M'] for e in valid_entries])[~outlier_mask], 
            c='blue', edgecolors='k', label='Γ→M', alpha=0.6)
plt.xlabel("Filling factor")
plt.ylabel("$v_g^\Gamma$")
plt.title("Group velocity vs filling factor")
plt.legend(frameon=True)
plt.grid(True)
plt.show()

# Isotropy Metric Plot
plt.figure(figsize=ROW1_COL1_SIZE)
plt.scatter(np.array(ff_values)[~outlier_mask], 
            np.array([e['isotropy'] for e in valid_entries])[~outlier_mask], 
            c='green', edgecolors='k')
plt.xlabel("Filling factor")
plt.ylabel("Isotropy Metric")
plt.title("Dirac Cone Isotropy")
plt.grid(True)
plt.show()

```

```python
import numpy as np
import matplotlib.pyplot as plt
import os
from scipy.optimize import curve_fit
from scipy.special import erf

# --- 1. Define the Skew-Gaussian ---
def skew_gaussian(x, A, mu, sigma, alpha):
    norm = (x - mu) / sigma
    gauss = np.exp(-(norm ** 2) / 2.0)
    skew = 0.5 * (1 + erf(alpha * norm / np.sqrt(2)))
    return A * gauss * skew * 2

# --- 2. Metric Computation Functions ---
# (Assuming compute_band6_polyfit, linearity_metric_v1, and linearity_metric_v2 remain unchanged from your code)

# --- 3. Process the Data ---
valid_entries = []
for entry in sims_lin:
    if entry["is_valid"]:
        compute_band6_polyfit(entry)
        # Average metrics across both directions
        v1_X = linearity_metric_v1(entry, "X")
        v1_M = linearity_metric_v1(entry, "M")
        entry["linearity_metric_v1"] = 0.5 * (v1_X + v1_M)
        
        v2_X = linearity_metric_v2(entry, delta_k=0.05, direction="X")
        v2_M = linearity_metric_v2(entry, delta_k=0.05, direction="M")
        entry["linearity_metric_v2"] = 0.5 * (v2_X + v2_M)
        valid_entries.append(entry)

# --- Extract FULL arrays (Length N) ---
ff_full = np.array([e["ff"] for e in valid_entries])
lm_v1_full = np.array([e["linearity_metric_v1"] for e in valid_entries])

# --- Create CLEAN arrays for fitting (Length N - Outliers) ---
ff_clean = ff_full[~outlier_mask]
lm_v1_clean = lm_v1_full[~outlier_mask]

# Mask for the fit range
ff_cut_left = 0.3
ff_cut_right = 0.55
fit_mask = (ff_clean > ff_cut_left) & (ff_clean < ff_cut_right)
x_fit = np.linspace(min(ff_full), max(ff_full), 100)

# --- 4. First Skew-Gaussian Fit (Metric 1) ---
p0_guesses = [lm_v1_clean.max(), ff_clean.mean(), 0.05, -2.0]
popt, _ = curve_fit(skew_gaussian, ff_clean[fit_mask], lm_v1_clean[fit_mask], p0=p0_guesses, maxfev=5000)
A_fit, mu_fit, sigma_fit, alpha_fit = popt
print(rf"Metric 1 Fit (1/m*): A={A_fit:.2f}, mu={mu_fit:.4f}, sigma={sigma_fit:.4f}, alpha={alpha_fit:.2f}")

# --- 5. Plotting Subplots ---
plt.figure(figsize=ROW1_COL2_SIZE)
y_fit = skew_gaussian(x_fit, *popt)
true_max = np.max(y_fit)

# --- Left Subplot: Inverse Effective Mass (1/m*) ---
plt.subplot(1, 2, 1)
# Plot clean data
plt.scatter(ff_clean, lm_v1_clean / true_max, c="red", edgecolors="k", s=20, label="Valid")
plt.plot(x_fit, y_fit / true_max, "b--", label=rf"$\mu={mu_fit:.3f}$, $\sigma={sigma_fit:.3f}$," + "\n" + rf" $\alpha={alpha_fit:.1f}$")

xlims = plt.xlim()
plt.xlim(xlims[0], xlims[1]*1.25) 
plt.ylim(-0.1, 1.3) 

ymin, ymax = plt.ylim()
plt.vlines(ff_cut_left, ymin, ymax, colors="gray", linestyles="--", label="Fit range")
plt.vlines(ff_cut_right, ymin, ymax, colors="gray", linestyles="--")
plt.xlabel("Filling factor")
plt.ylabel("1/m* (Direction Averaged)")
plt.title(r"Linearity Metric 1")
plt.grid(True, alpha=0.5)
plt.legend(loc="best", fontsize=7, frameon=True)

# --- Right Subplot: Normalized Metric v2 ---
plt.subplot(1, 2, 2)
dks = [0.05, 0.005]

for i, delta_k in enumerate(dks):
    # Calculate full array dynamically
    lm_v2_full = np.array([0.5 * (linearity_metric_v2(e, delta_k, "X") + linearity_metric_v2(e, delta_k, "M")) for e in valid_entries])
    # Create clean array for fitting
    lm_v2_clean = lm_v2_full[~outlier_mask]

    p0_v2 = [lm_v2_clean.max(), ff_clean.mean(), 0.05, -2.0]
    popt_v2, _ = curve_fit(skew_gaussian, ff_clean, lm_v2_clean, p0=p0_v2, maxfev=5000)
    A_v2, mu_v2, sigma_v2, alpha_v2 = popt_v2
    print(rf"\delta k={delta_k:.3f} Fit: A={A_v2:.2f}, mu={mu_v2:.4f}, sigma={sigma_v2:.4f}, alpha={alpha_v2:.2f}")

    y_fit_v2_full = skew_gaussian(x_fit, *popt_v2)
    true_max_v2 = np.max(y_fit_v2_full)

    # Plot clean data and fit
    plt.scatter(ff_clean, lm_v2_clean / true_max_v2, edgecolors="k", s=20, label=rf"$\Delta k={delta_k}$ Valid")
    plt.plot(x_fit, y_fit_v2_full / true_max_v2, "--", label=rf"$\mu={mu_v2:.3f}$, $\sigma={sigma_v2:.3f}$" + "\n" + rf"$\alpha={alpha_v2:.1f}$")
    
    

xlims = plt.xlim()
plt.xlim(xlims[0], xlims[1]*1.2) 
plt.ylim(-0.1, 1.2) 
plt.xlabel("Filling factor")
plt.ylabel(r"LIN : NLIN")
plt.title("Linearity Metric 2")
plt.grid(True, alpha=0.5)
plt.legend(loc="upper right", fontsize=7, frameon=True)
plt.tight_layout()

if 'figure_path' in globals():
    plt.savefig(os.path.join(figure_path, "linearity_metrics_vs_ff.png"))   
plt.show()

# --- 6. Vectorized (r1, r2) Space Plot ---
plt.figure(figsize=(4.2, 3.5))

invalid_sims = [e for e in sims_lin if not e["is_valid"]]
if invalid_sims:
    r1_inv = [e["r1"] for e in invalid_sims]
    r2_inv = [e["r2"] for e in invalid_sims]
    plt.scatter(r1_inv, r2_inv, c="lightgray", edgecolors="k", s=20, label="Invalid")

r1_full = np.array([e["r1"] for e in valid_entries], dtype=float)
r2_full = np.array([e["r2"] for e in valid_entries], dtype=float)

# Recalculate size mapping based on the base delta_k=0.05
lm_v2_base = np.array([e["linearity_metric_v2"] for e in valid_entries], dtype=float)
s_min, s_max = 10.0, 50.0
v2_min, v2_max = np.min(lm_v2_base[~outlier_mask]), np.max(lm_v2_base[~outlier_mask]) # Size scale relative to clean data
if np.isclose(v2_max, v2_min):
    sizes_full = np.full_like(lm_v2_base, (s_min + s_max) / 2.0)
else:
    sizes_full = s_min + (lm_v2_base - v2_min) * (s_max - s_min) / (v2_max - v2_min)
    sizes_full = np.clip(sizes_full, s_min, s_max) # Ensure outliers don't blow up the size scale

# Plot clean points
sc = plt.scatter(r1_full[~outlier_mask], r2_full[~outlier_mask], c=lm_v1_full[~outlier_mask], 
                 s=sizes_full[~outlier_mask], cmap="viridis", edgecolors="k")

# Plot outlier points
plt.scatter(r1_full[outlier_mask], r2_full[outlier_mask], c='red', 
            s=sizes_full[outlier_mask],  label='Outliers', marker='x')

plt.colorbar(sc, label=r"Metric v1 (1/m*) Avg")

plt.xlabel("$r_1/a$")
plt.ylabel("$r_2/a$")
plt.title(r"Linearity Metrics in $(r_1, r_2)$ Space" + "\n Color: 1/m*, Size: LIN/NLIN Ratio")
if invalid_sims:
    plt.legend(loc="best", fontsize=8)
plt.grid(True, alpha=0.3)
plt.tight_layout()

if 'figure_path' in globals():
    plt.savefig(os.path.join(figure_path, "linearity_metrics_vs_r1_r2.png"))
plt.show()

from IPython.display import display, Math
formula = r"\text{Linearity Metric v2} = \frac{|a_1 \Delta k|}{|a_2 \Delta k^2| + |a_3 \Delta k^3| + \epsilon}"
display(Math(formula))
```

```python
### Raw data
# ===========================================================================
# Filling Factor vs Concentration Factor, dot size ∝ Linearity Metric 2
# ===========================================================================
# CF(band) = ∫_diel ε|E|² / ∫_all ε|E|²   at Γ
# ===========================================================================
import json

cf_cache_path = os.path.join(data_rooth, f"{name_2D_optimized}_cf_lin_cache.json")

# --- Gamma-only config WITH E-field output ---
config_cf_lin = dict(
    resolution=RESOLUTION,
    num_bands=8,
    k_points=[mp.Vector3(0, 0, 0)],
    extra_runner_command="fix-hfield-phase output-efield output-hfield",
)
mpb_config_cf_lin = MPBSchemeConfigurator(photonic_crystal_2D, ["te"], **config_cf_lin)
script_cf_lin = mpb_config_cf_lin.get_scheme_config(join_newline=True)

# --- One Simulation per valid sims_lin entry ---
valid_entries_cf = [e for e in sims_lin if e["is_valid"]]
sims_cf_lin = []
for j, entry in enumerate(valid_entries_cf):
    sim_name = f"{name_2D_optimized}_cf_lin_{j:03d}"
    sim_cf = Simulation(
        simulation_name=sim_name,
        script=script_cf_lin,
        directory=os.path.join(data_rooth, sim_name),
    )
    sims_cf_lin.append({
        'r1': entry['r1'], 'r2': entry['r2'], 'sim': sim_cf,
        'entry': entry,
    })

# --- Run (fast – single k-point) ---
if SIMULATE:
    print(f"Running {len(sims_cf_lin)} Γ-only simulations with E-field output …")
    def run_cf_sim(args):
        j, s = args
        s['sim'].run_hpc(
            mpb_command_line_params=dict(r1=s['r1'], r2=s['r2']),
            mpi=True, load_epsilon=False)
        return j

    with ThreadPoolExecutor(max_workers=N_WORKERS) as pool:
        futures = {pool.submit(run_cf_sim, (j, s)): j for j, s in enumerate(sims_cf_lin)}
        for fut in as_completed(futures):
            j = fut.result()
            print(f"  [{j+1}/{len(sims_cf_lin)}] r1={sims_cf_lin[j]['r1']:.4f}, "
                  f"r2={sims_cf_lin[j]['r2']:.4f} done")
    print("All CF-linearity simulations completed.")
else:
    print("SIMULATE=False → loading existing CF data")

# --- Compute or load concentration factors for bands 4, 5, 6 ---
bands_cf_lin = [4, 5, 6]
opt_cf_lin = MPBDataOptions(rectify=True, periods=1)

if os.path.exists(cf_cache_path) and not SIMULATE:
    # Load cached CF data
    with open(cf_cache_path, 'r') as f:
        cf_cache = json.load(f)
    cf_lin_data = {int(k): v for k, v in cf_cache.items()}
    print(f"Loaded cached CF data from {cf_cache_path} ({len(cf_lin_data[bands_cf_lin[0]])} entries)")
else:
    # Compute CF from simulation field data
    cf_lin_data = {b: [] for b in bands_cf_lin}

    for j, s in enumerate(sims_cf_lin):
        sim = s['sim']
        try:
            eps = sim.load_and_convert_epsilon_data(options=opt_cf_lin)
            if eps.ndim == 3:
                eps = eps[:, :, eps.shape[2] // 2]

            eps_mid = (1.0 + eps.max()) / 2.0
            diel_mask = eps > eps_mid

            for band in bands_cf_lin:
                Ex = sim.load_and_convert_field_data(
                    1, band, "x", "te", "e", conversion_options=opt_cf_lin)
                Ey = sim.load_and_convert_field_data(
                    1, band, "y", "te", "e", conversion_options=opt_cf_lin)
                if Ex.ndim == 3:
                    Ex = Ex[:, :, Ex.shape[2] // 2]
                if Ey.ndim == 3:
                    Ey = Ey[:, :, Ey.shape[2] // 2]

                eps_Esq = eps * (np.abs(Ex)**2 + np.abs(Ey)**2)
                total = eps_Esq.sum()
                cf = float(eps_Esq[diel_mask].sum() / total) if total > 0 else None
                cf_lin_data[band].append(cf)

        except Exception as e:
            print(f"  Error j={j}, r1={s['r1']:.4f}, r2={s['r2']:.4f}: {e}")
            for band in bands_cf_lin:
                cf_lin_data[band].append(None)

    # Save to cache
    cf_cache_save = {str(k): v for k, v in cf_lin_data.items()}
    with open(cf_cache_path, 'w') as f:
        json.dump(cf_cache_save, f)
    print(f"Saved CF data to {cf_cache_path}")

    # Convert None → NaN for numpy compatibility
    cf_lin_data = {b: [x if x is not None else float('nan') for x in vals]
                   for b, vals in cf_lin_data.items()}

# Store CF back into valid entries
for j, s in enumerate(sims_cf_lin):
    for band in bands_cf_lin:
        s['entry'][f'cf_band_{band}'] = cf_lin_data[band][j]

# --- Compute linearity metric v2 for chosen delta_k ---
delta_k_plot = 0.001
lm_v2_vals = np.array([linearity_metric_v2(s['entry'], delta_k=delta_k_plot) for s in sims_cf_lin])
# Normalize by max for better plotting
lm_v2_vals /= lm_v2_vals.max()
ff_cf = np.array([s['entry']['ff'] for s in sims_cf_lin])

print(f"\nConcentration-factor statistics (N = {len(sims_cf_lin)}, Δk = {delta_k_plot}):")
for band in bands_cf_lin:
    v = np.array(cf_lin_data[band], dtype=float)
    v = v[~np.isnan(v)]
    if len(v) > 0:
        print(f"  Band {band}: CF = {v.mean():.4f} ± {v.std():.4f}  [{v.min():.4f}, {v.max():.4f}]")

# ===========================================================================
# 2D Plot: x = Filling Factor, y = Concentration Factor
#          dot size ∝ Linearity Metric 2,  dot color = band
# ===========================================================================
band_colors = {4: 'tab:blue', 5: 'tab:orange', 6: 'tab:green'}
band_markers = {4: 'o', 5: 's', 6: 'D'}

# Normalize LM2 to dot sizes
size_min_lm, size_max_lm = 5, 50
lm_min, lm_max = lm_v2_vals.min(), lm_v2_vals.max()
if lm_max > lm_min:
    lm_norm = (lm_v2_vals - lm_min) / (lm_max - lm_min)
else:
    lm_norm = np.ones_like(lm_v2_vals) * 0.5
dot_sizes_lm = size_min_lm + (size_max_lm - size_min_lm) * lm_norm

fig, ax = plt.subplots( constrained_layout=True)

for band in [4,5,6]:
    cf_vals = np.array(cf_lin_data[band], dtype=float)
    valid_mask = ~np.isnan(cf_vals)
    ax.scatter(ff_cf[valid_mask], cf_vals[valid_mask],
               s=dot_sizes_lm[valid_mask],
               color=band_colors[band], marker=band_markers[band],
               edgecolors='k', linewidth=0.5, alpha=0.8, zorder=3,
               label=f'Band {band}')

ax.set_xlabel('Filling factor')
ax.set_ylabel(r'Concentration factor')
ax.set_title(f'Linearity Metric')
ax.grid(True, alpha=0.3)

# Size legend for LM2 values
for lm_leg in [lm_min, 0.5 * (lm_min + lm_max), lm_max]:
    s_leg = size_min_lm + (size_max_lm - size_min_lm) * (
        (lm_leg - lm_min) / (lm_max - lm_min) if lm_max > lm_min else 0.5)
    ax.scatter([], [], s=s_leg, c='gray', edgecolors='k', linewidth=0.5,
               label=f'LM2 = {lm_leg:.1f}')

ax.legend(loc='best', fontsize=7)
plt.savefig(os.path.join(figure_path, "linearity_metric_v2_vs_CF.png"))
plt.show()


# Latex display of the confinement factor formula
from IPython.display import display, Math
formula = r"""
\text{CF} = \frac{\int_{\text{dielectric}} \varepsilon |\mathbf{E}|^2 \, dV}{\int_{\text{all}} \varepsilon |\mathbf{E}|^2 \, dV}
"""
display(Math(formula))

# In a new figure, plot the frequencies at gamma for bands 4, 5, 6 against filling factor, with dot size ∝ LM2
# Subtract the frequency of band 5 at gamma from all bands
fig, ax = plt.subplots(figsize=ROW1_COL1_SIZE)

# Get band 5 frequencies at gamma for reference
freqs_band5_gamma = np.array([s['entry']['bands'][f'band_5'][s['entry']['gamma_idx']] for s in sims_cf_lin], dtype=float)

for band in bands_cf_lin:
    freqs_gamma = np.array([s['entry']['bands'][f'band_{band}'][s['entry']['gamma_idx']] for s in sims_cf_lin], dtype=float)
    # Subtract band 5 frequency
    freqs_gamma_diff = freqs_gamma - freqs_band5_gamma
    valid_mask = ~np.isnan(freqs_gamma_diff)
    ax.scatter(ff_cf[valid_mask], freqs_gamma_diff[valid_mask],
               s=dot_sizes_lm[valid_mask],
               color=band_colors[band], marker=band_markers[band],
               edgecolors='k', linewidth=0.5, alpha=0.8, zorder=3,
               label=f'Band {band}')
ax.set_xlabel('Filling factor')
ax.set_ylabel(r'Frequency at Γ - Band 5 freq. (ωa/2πc)')
ax.set_title('Relative Frequency at Γ vs Filling Factor')
ax.grid(True, alpha=0.3)
ax.legend(loc='best', fontsize=7)
plt.savefig(os.path.join(figure_path, "gamma_frequencies_vs_ff.png"))
plt.show()

# Plot Hz for the bands with FF closest to 0.8 
target_ff = 0.38
closest_idx = np.argmin(np.abs(ff_cf - target_ff))
sim_closest = sims_cf_lin[closest_idx]['sim']
fig, axs = plt.subplots(1,3, figsize=ROW1_COL3_SIZE)
for band, ax in zip(bands_cf_lin, axs):
    Hz = sim_closest.load_and_convert_field_data(
        1, band, "z", "te", "h", conversion_options=opt_cf_lin)
    if Hz.ndim == 3:
        Hz = Hz[:, :, Hz.shape[2] // 2]
    im = ax.imshow(np.abs(Hz.T), origin='lower', cmap='inferno')
    ax.set_title(f'|Hz| for Band {band} at FF ≈ {ff_cf[closest_idx]:.3f}')
plt.colorbar(im, ax=ax, label='|Hz|')
plt.savefig(os.path.join(figure_path, f"Hz_band_{band}_ff_{ff_cf[closest_idx]:.3f}.png"))
plt.show()
```

```python
## Filing factor vs concentration factor
# ===========================================================================
# Filling Factor vs Concentration Factor (Subspace Reconstruction & Irrep ID)
# Dot size ∝ Linearity Metric 2, Dot color = Reconstructed Irrep
# ===========================================================================
import json
import os
import numpy as np
import matplotlib.pyplot as plt
from concurrent.futures import ThreadPoolExecutor, as_completed
from IPython.display import display, Math

# --- Symmetry Projection & Classification Functions ---
def unmix_E_field(Ex_mixed, Ey_mixed):
    """Separates E-fields into 1D (A1/B1/etc) and 2D (E) symmetry components."""
    Ex_flip = np.rot90(Ex_mixed, k=2, axes=(0, 1))
    Ey_flip = np.rot90(Ey_mixed, k=2, axes=(0, 1))
    
    # If a field is purely 1D (A1/B1), it will flip sign under C4, so we take the difference to isolate it.
    # E(x,y) = -E(-y,x) for 1D irreps, while E(x,y) = E(-y,x) for the 2D E irrep.
    Ex_1D = 0.5 * (Ex_mixed - Ex_flip)
    Ey_1D = 0.5 * (Ey_mixed - Ey_flip)
    
    Ex_E = 0.5 * (Ex_mixed + Ex_flip)
    Ey_E = 0.5 * (Ey_mixed + Ey_flip)
    
    return (Ex_1D, Ey_1D), (Ex_E, Ey_E)

def unmix_Hz_field(Hz_mixed):
    """Separates pseudo-scalar Hz into 1D and 2D symmetry components."""
    Hz_flip = np.rot90(Hz_mixed, k=2, axes=(0, 1))
    Hz_1D = 0.5 * (Hz_mixed + Hz_flip)
    Hz_E  = 0.5 * (Hz_mixed - Hz_flip)
    return Hz_1D, Hz_E

def classify_1D_irrep(Ex, Ey):
    """Identifies the specific 1D irrep (A1, A2, B1, B2) using overlap integrals."""
    norm = np.sum(np.abs(Ex)**2 + np.abs(Ey)**2)
    if norm < 1e-10:
        return "Unknown 1D"
        
    # 1. Evaluate C4 (90-degree) rotation overlap
    Ex_C4 = -np.rot90(Ey, k=1, axes=(0, 1))
    Ey_C4 = np.rot90(Ex, k=1, axes=(0, 1))
    overlap_C4 = np.real(np.sum(np.conj(Ex) * Ex_C4 + np.conj(Ey) * Ey_C4) / norm)
    
    # 2. Evaluate Sigma_v (mirror across y-axis: x -> -x) overlap
    Ex_sig = -np.flip(Ex, axis=0)
    Ey_sig = np.flip(Ey, axis=0)
    overlap_sig = np.real(np.sum(np.conj(Ex) * Ex_sig + np.conj(Ey) * Ey_sig) / norm)
    
    # Map the characters (+1 or -1) to the C4v Irreps
    if overlap_C4 > 0.5:
        return "A1" if overlap_sig > 0.5 else "A2"
    else:
        return "B1" if overlap_sig > 0.5 else "B2"

# --- Simulation Setup ---

cf_cache_path = os.path.join(data_rooth, f"{name_2D_optimized}_cf_lin_cache_unmixed.json")

config_cf_lin = dict(
    resolution=RESOLUTION,
    num_bands=8,
    k_points=[mp.Vector3(0, 0, 0)],
    extra_runner_command="fix-hfield-phase output-efield output-hfield",
)
mpb_config_cf_lin = MPBSchemeConfigurator(photonic_crystal_2D, ["te"], **config_cf_lin)
script_cf_lin = mpb_config_cf_lin.get_scheme_config(join_newline=True)

valid_entries_cf = [e for e in sims_lin if e["is_valid"]]
sims_cf_lin = []
for j, entry in enumerate(valid_entries_cf):
    sim_name = f"{name_2D_optimized}_cf_lin_{j:03d}"
    sim_cf = Simulation(simulation_name=sim_name, script=script_cf_lin, directory=os.path.join(data_rooth, sim_name))
    sims_cf_lin.append({'r1': entry['r1'], 'r2': entry['r2'], 'sim': sim_cf, 'entry': entry})

# --- Run (fast – single k-point) ---
if SIMULATE:
    print(f"Running {len(sims_cf_lin)} Γ-only simulations...")
    def run_cf_sim(args):
        j, s = args
        s['sim'].run_hpc(mpb_command_line_params=dict(r1=s['r1'], r2=s['r2']), mpi=True, load_epsilon=False)
        return j

    with ThreadPoolExecutor(max_workers=N_WORKERS) as pool:
        futures = {pool.submit(run_cf_sim, (j, s)): j for j, s in enumerate(sims_cf_lin)}
        for fut in as_completed(futures):
            j = fut.result()
            print(f"  [{j+1}/{len(sims_cf_lin)}] r1={sims_cf_lin[j]['r1']:.4f} done")

# --- Compute Subspace-Reconstructed CFs & True Uncoupled Frequencies ---
bands_cf_lin = [4, 5, 6]
opt_cf_lin = MPBDataOptions(rectify=True, periods=1)

# Dynamic dictionary to handle whatever 1D irrep is found
irrep_data = {
    'A1 Mode': {'ff': [], 'cf': [], 'band': []},
    'A2 Mode': {'ff': [], 'cf': [], 'band': []},
    'B1 Mode': {'ff': [], 'cf': [], 'band': []},
    'B2 Mode': {'ff': [], 'cf': [], 'band': []},
    'E Mode':  {'ff': [], 'cf': [], 'band': []}
}

for j, s in enumerate(sims_cf_lin):
    sim = s['sim']
    
    rho_1D_total = 0
    rho_E_total = 0
    W_1D_total = 0
    W_E_total = 0
    freq_weighted_1D = 0
    freq_weighted_E = 0
    
    best_1D_weight = -1
    best_Ex_1D, best_Ey_1D = None, None
    
    try:
        eps = sim.load_and_convert_epsilon_data(options=opt_cf_lin)
        if eps.ndim == 3: eps = eps[:, :, eps.shape[2] // 2]
        diel_mask = eps > ((1.0 + eps.max()) / 2.0)

        for band in bands_cf_lin:
            Ex = sim.load_and_convert_field_data(1, band, "x", "te", "e", conversion_options=opt_cf_lin)
            Ey = sim.load_and_convert_field_data(1, band, "y", "te", "e", conversion_options=opt_cf_lin)
            if Ex.ndim == 3: Ex = Ex[:, :, Ex.shape[2] // 2]
            if Ey.ndim == 3: Ey = Ey[:, :, Ey.shape[2] // 2]

            (Ex_1D, Ey_1D), (Ex_E, Ey_E) = unmix_E_field(Ex, Ey)
            
            rho_1D_band = np.abs(Ex_1D)**2 + np.abs(Ey_1D)**2
            rho_E_band  = np.abs(Ex_E)**2 + np.abs(Ey_E)**2
            
            rho_1D_total += rho_1D_band
            rho_E_total  += rho_E_band
            
            freq = s['entry']['bands'][f'band_{band}'][s['entry']['gamma_idx']]
            W_1D = np.sum(rho_1D_band)
            W_E  = np.sum(rho_E_band)
            
            # Save the cleanest 1D projection for precise irrep identification
            if W_1D > best_1D_weight:
                best_1D_weight = W_1D
                best_Ex_1D, best_Ey_1D = Ex_1D.copy(), Ey_1D.copy()
            
            W_1D_total += W_1D
            W_E_total += W_E
            freq_weighted_1D += W_1D * freq
            freq_weighted_E  += W_E * freq

        # Identify specific 1D Irrep
        mode_1D_label = f"{classify_1D_irrep(best_Ex_1D, best_Ey_1D)} Mode"

        # Calculate pure Concentration Factors
        eps_Esq_1D = eps * rho_1D_total
        total_1D = eps_Esq_1D.sum()
        cf_1D = float(eps_Esq_1D[diel_mask].sum() / total_1D) if total_1D > 0 else float('nan')
        
        eps_Esq_E = eps * rho_E_total
        total_E = eps_Esq_E.sum()
        cf_E = float(eps_Esq_E[diel_mask].sum() / total_E) if total_E > 0 else float('nan')

        irrep_data[mode_1D_label]['ff'].append(s['entry']['ff'])
        irrep_data[mode_1D_label]['cf'].append(cf_1D)
        irrep_data[mode_1D_label]['band'].append(j)

        irrep_data['E Mode']['ff'].append(s['entry']['ff'])
        irrep_data['E Mode']['cf'].append(cf_E)
        irrep_data['E Mode']['band'].append(j)

        # Calculate true uncoupled frequencies and cost
        if W_1D_total > 0 and W_E_total > 0:
            f_1D_true = freq_weighted_1D / W_1D_total
            f_E_true  = freq_weighted_E / W_E_total
            s['entry']['cost'] = abs(f_1D_true - f_E_true)
        else:
            s['entry']['cost'] = float('nan')

    except Exception as e:
        print(f"  Error j={j}: {e}")
        s['entry']['cost'] = float('nan')

# --- Compute Linearity for Dot Sizes ---
delta_k_plot = 0.001
lm_v2_vals = np.array([linearity_metric_v2(s['entry'], delta_k=delta_k_plot) for s in sims_cf_lin])
lm_max = lm_v2_vals.max()
lm_min = lm_v2_vals.min()
lm_norm = (lm_v2_vals - lm_min) / (lm_max - lm_min) if lm_max > lm_min else np.ones_like(lm_v2_vals) * 0.5

size_min_lm, size_max_lm = 5, 50
dot_sizes_lm = size_min_lm + (size_max_lm - size_min_lm) * lm_norm

# ===========================================================================
# Plot 1: Concentration Factor by Reconstructed Irrep
# ===========================================================================
fig, ax = plt.subplots(constrained_layout=True)

irrep_styles = {
    'A1 Mode': {'color': 'tab:blue', 'marker': 'o'},
    'A2 Mode': {'color': 'tab:cyan', 'marker': 'o'},
    'B1 Mode': {'color': 'tab:purple', 'marker': 'o'},
    'B2 Mode': {'color': 'tab:red', 'marker': 'o'},
    'E Mode':  {'color': 'tab:orange', 'marker': 's'}
}

for irrep, style in irrep_styles.items():
    if len(irrep_data[irrep]['ff']) == 0: continue # Skip empty irrep lists
        
    ff_vals = np.array(irrep_data[irrep]['ff'])
    cf_vals = np.array(irrep_data[irrep]['cf'])
    indices = np.array(irrep_data[irrep]['band'])
    
    valid_mask = ~np.isnan(cf_vals)
    sizes = np.array([dot_sizes_lm[idx] for idx in indices])
    
    ax.scatter(ff_vals[valid_mask], cf_vals[valid_mask], s=sizes[valid_mask],
               color=style['color'], marker=style['marker'],
               edgecolors='k', linewidth=0.5, alpha=0.8, zorder=3, label=irrep)

ax.set_xlabel('Filling factor')
ax.set_ylabel('Concentration factor (Reconstructed)')
ax.set_title('Linearity Metric (Classified by Irrep)')
ax.grid(True, alpha=0.3)

for lm_leg in [lm_min, 0.5 * (lm_min + lm_max), lm_max]:
    s_leg = size_min_lm + (size_max_lm - size_min_lm) * (
        (lm_leg - lm_min) / (lm_max - lm_min) if lm_max > lm_min else 0.5)
    ax.scatter([], [], s=s_leg, c='gray', edgecolors='k', linewidth=0.5, label=f'LM2 = {lm_leg:.1f}')

ax.legend(loc='best', fontsize=7)
plt.savefig(os.path.join(figure_path, "linearity_metric_v2_vs_CF_reconstructed.png"))
plt.show()

# ===========================================================================
# Plot 2: FOM (log(1/cost))
# ===========================================================================
costs = np.array([s['entry'].get('cost', float('nan')) for s in sims_cf_lin])
valid_costs = (costs > 0) & (~np.isnan(costs))
fom = np.zeros_like(costs)
fom[valid_costs] = np.log(1.0 / costs[valid_costs])

fig, ax = plt.subplots()
ax.scatter(np.array([s['entry']['ff'] for s in sims_cf_lin])[valid_costs], 
           fom[valid_costs], color='purple', edgecolors='k', alpha=0.8)
ax.set_xlabel('Filling factor')
ax.set_ylabel(r'$\log(1 / \text{cost})$')
ax.set_title('Dirac Cone FOM vs Filling Factor')
ax.grid(True, alpha=0.3)
plt.savefig(os.path.join(figure_path, "dirac_cone_fom_vs_ff.png"))
plt.show()

# ===========================================================================
# Plot 3: Unmixed |Hz| Fields at target FF
# ===========================================================================

target_ff = 0.38
ff_array = np.array([s['entry']['ff'] for s in sims_cf_lin])
closest_idx = np.argmin(np.abs(ff_array - target_ff))
sim_closest = sims_cf_lin[closest_idx]['sim']

fig, axs = plt.subplots(1, 3, figsize=(15, 5))
for band, ax in zip(bands_cf_lin, axs):
    Hz = sim_closest.load_and_convert_field_data(1, band, "z", "te", "h", conversion_options=opt_cf_lin)
    if Hz.ndim == 3: Hz = Hz[:, :, Hz.shape[2] // 2]
    
    # Project the field
    Hz_1D, Hz_E = unmix_Hz_field(Hz)
    
    # Auto-detect which symmetry belongs to this band index
    if np.sum(np.abs(Hz_1D)**2) > np.sum(np.abs(Hz_E)**2):
        Hz_pure = Hz_1D
        label = "1D Mode"
    else:
        Hz_pure = Hz_E
        label = "E Mode"

    im = ax.imshow(np.real(Hz_pure.T), origin='lower', cmap='RdBu')
    ax.set_title(f'Hz (Band {band})\n{label} at FF ≈ {ff_array[closest_idx]:.3f}')
    plt.colorbar(im, ax=axs, label='Hz')

plt.tight_layout()
plt.savefig(os.path.join(figure_path, f"Hz_unmixed_ff_{ff_array[closest_idx]:.3f}.png"))
plt.show()
```

#### Functions

```python
## Calculate effective epsilon and mu from field data

from phc_nzi.simulation_handler import Simulation
from phc_nzi.simulation_viewer import SimulationViewer
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import numpy as np
from phc_nzi.simulation_handler import MPBDataOptions, MPBDataConverter
import concurrent.futures







class FieldAnalyzer:
    def __init__(self,simulation: Simulation,  bands, polarization, k_direction):
        self.bands = bands
        self.polarization = polarization
        self.simulation = simulation 
        self.df = self.simulation.load_frequency_data(self.polarization)  
        self.result_df = None
        self.k_direction = k_direction

    def load_field_data(self, k_index, band_index, field_type, nonbloch = True, comp = "z", conversion_options: MPBDataOptions = None, overwrite=False):
        field =  self.simulation.load_and_convert_field_data(k_index, band_index, comp, self.polarization, field_type, 
                                                           nonbloch = nonbloch, conversion_options=conversion_options,
                                                           file_comp=comp, overwrite=overwrite)  
        if field is None:
            raise ValueError("Field data is None")
        return self.make_2D(field)
    
    def load_hfield_data(self, k_index, band_index, comp, nonbloch = True, conversion_options: MPBDataOptions = None, overwrite=False): 
        return self.load_field_data(k_index, band_index, "h", nonbloch, comp, conversion_options, overwrite=overwrite)
    
    def load_efield_data(self, k_index, band_index, comp, nonbloch = True, conversion_options: MPBDataOptions = None, overwrite=False): 
        return self.load_field_data(k_index, band_index, "e", nonbloch, comp ,conversion_options, overwrite=overwrite)    
    
    def make_2D(self, field_data):
        if field_data.ndim == 2:
            return field_data
        elif field_data.ndim == 3:
            return field_data[:,:,field_data.shape[2]//2]
        else: 
            raise ValueError("Field data must be 2D or 3D")
        
    def field_less_than_zero(self, field_data):
        """Returns a boolean mask for values less than zero"""
        return field_data < 0
    
    def field_greater_than_zero(self, field_data):
        """Returns a boolean mask for values greater than zero or equal to zero"""
        return field_data >= 0
    
    
    def get_mask(self, field_data, mask_func):
        return mask_func(field_data)
    
    def mask_field_data(self, field_data, mask):
        """Mask field data by setting values where mask is False to None/np.nan
        while preserving the shape of the array"""
        masked = field_data.copy()
        masked[~mask] = 0
        return masked

    def average_2D(self, field_data):
        if field_data.ndim == 3:
            raise ValueError("Field data must be 2D")
        shape = field_data.shape
        return field_data.sum() / (shape[0] * shape[1])
    
    def average_1D(self, field_data, axis):
        if field_data.ndim == 3:
            raise ValueError("Field data must be 2D")
        if axis not in [0, 1]:
            raise ValueError("Axis must be 0 or 1")
        if axis == 0:
            return field_data[0,:].sum() / field_data.shape[1]
        elif axis == 1:
            return field_data[:,0].sum() / field_data.shape[0]
        
    def get_kmag_and_freq(self, k_idx, b_idx):
        kmag, freq = self.simulation.get_kmag_and_freq(self.df, k_idx,  b_idx, self.polarization,)
        return kmag, freq


    
    def get_freq(self, k_idx, b_idx):
        return self.simulation.get_freq(self.df, k_idx, b_idx, self.polarization,)

    

    def get_eps_mu(self, k_index, band_index, component_i, component_j, average_type = "2D", nonbloch = True, conversion_options: MPBDataOptions = None, overwrite=True):
        e_field_i = self.load_efield_data(k_index, band_index, component_i,  nonbloch = nonbloch, conversion_options=conversion_options, overwrite=overwrite)
        h_field_j = self.load_hfield_data(k_index, band_index, component_j,  nonbloch = nonbloch, conversion_options=conversion_options, overwrite=overwrite)
        
        if self.k_direction == "x":
            if self.polarization == "tm" or self.polarization == "zodd":
                sgn_mu = 1
                sgn_eps = 1

            elif self.polarization == "te" or self.polarization == "zeven":
                sgn_mu = 1
                sgn_eps = -1
        if self.k_direction == "y":
            if self.polarization == "tm" or self.polarization == "zodd":
                sgn_mu = -1
                sgn_eps = -1

            elif self.polarization == "te" or self.polarization == "zeven":
                sgn_mu = -1
                sgn_eps = 1
        
        mask_I = self.get_mask(h_field_j.real, self.field_greater_than_zero)
        mask_II = self.get_mask(h_field_j.real, self.field_less_than_zero)

        # Create the masked complex fields for both regions
        e_field_i_masked_I = self.mask_field_data(e_field_i, mask_I)
        e_field_i_masked_II = self.mask_field_data(e_field_i, mask_II)
        h_field_j_masked_I = self.mask_field_data(h_field_j, mask_I)
        h_field_j_masked_II = self.mask_field_data(h_field_j, mask_II)

        if average_type == "2D":
            e_field_i_masked_I_avg = self.average_2D(e_field_i_masked_I)
            e_field_i_masked_II_avg = self.average_2D(e_field_i_masked_II)
            h_field_j_masked_I_avg = self.average_2D(h_field_j_masked_I)
            h_field_j_masked_II_avg = self.average_2D(h_field_j_masked_II)

        elif average_type == "1D":
            e_field_i_masked_I_avg = self.average_1D(e_field_i_masked_I, 0)
            e_field_i_masked_II_avg = self.average_1D(e_field_i_masked_II, 0)
            h_field_j_masked_I_avg = self.average_1D(h_field_j_masked_I, 0)
            h_field_j_masked_II_avg = self.average_1D(h_field_j_masked_II, 0)


        e_over_h_I = e_field_i_masked_I_avg / h_field_j_masked_I_avg
        e_over_h_II = e_field_i_masked_II_avg / h_field_j_masked_II_avg

        e_over_h_I = e_over_h_I.real
        e_over_h_II = e_over_h_II.real  

        mu_I = e_over_h_I
        mu_II = e_over_h_II

        eps_I = 1/e_over_h_I
        eps_II = 1/e_over_h_II

        k_mag, freq = self.get_kmag_and_freq(k_index, band_index)
        k_mag_over_freq = k_mag / freq
        mu = np.emath.sqrt(mu_I)*np.emath.sqrt(mu_II)
        eps = np.emath.sqrt(eps_I)*np.emath.sqrt(eps_II)
        
        eps = eps*k_mag_over_freq
        mu = mu*k_mag_over_freq
        return  eps, mu, e_over_h_I, e_over_h_II
    
    def get_impedance(self, eps, mu):
        return np.emath.sqrt(mu/eps)
    
    def get_refractive_index(self, eps, mu):
        return np.emath.sqrt(eps)*np.emath.sqrt(mu)
    
    def get_k_indices(self):
        return self.simulation.get_kpoints_indices(self.df)
    
    def get_eps_mu_impedance_neff(self, component_i, component_j , average_type = "2D", nonbloch = True, conversion_options: MPBDataOptions = None, plot = False, overwrite=True):
        k_indices = self.get_k_indices()
        bands_freqs = []
        for band in self.bands:
            freqs = []
            for k_idx in k_indices:
                freq = self.get_freq(k_idx, band)
                freqs.append(freq)
            bands_freqs.append(freqs)
        if plot is True:
            plt.figure()
            plt.title("Selected Points")
            for idx, freqs in enumerate(bands_freqs):
                plt.plot(k_indices, freqs, "ro", label=f"Band {idx}")
                plt.xlabel("K-Index")
                plt.ylabel("Frequency")
                plt.grid(True)
            plt.legend()

        results = []
        requests = []
        for k_idx in k_indices:
            for band in self.bands:
                requests.append((self, k_idx, band, component_i, component_j, average_type, nonbloch, conversion_options, overwrite))
        
        # Parallel execution
        with concurrent.futures.ProcessPoolExecutor() as executor:
            futures = [executor.submit(_retrieve_single_point, *req) for req in requests]
            for future in concurrent.futures.as_completed(futures):
                try:
                    res = future.result()
                    results.append(res)
                    # print(f"Finished K-Index: {res['k_index']}, Band: {res['band']}") # Reduced verbosity for speed
                except Exception as e:
                    print(f"Error processing item: {e}")
        
        # Sort results because parallel execution might shuffle them
        results.sort(key=lambda x: (x['band'], x['k_index']))
        
        # Convert results to DataFrame
        results_df = pd.DataFrame(results)
        self.result_df = results_df
        return results_df

    
    
    def plot_eps_vs_freqs(self):
        results_df = self.result_df
        if results_df is None:
            print("No results found. Run get_eps_mu_impedance_neff() first.")
            return
        plt.plot(results_df["frequency"], results_df["eps"], "ro", label="$\\varepsilon_{eff}$")
        plt.xlabel("Frequency")
        plt.legend()
        plt.grid(True)
    
    def plot_mu_vs_freqs(self):
        results_df = self.result_df
        if results_df is None:
            print("No results found. Run get_eps_mu_impedance_neff() first.")
            return
        plt.plot(results_df["frequency"], results_df["mu"], "bo", label="$\\mu_{eff}$")
        plt.xlabel("Frequency")
        plt.legend()    
        plt.grid(True)

    def plot_impedance_vs_freqs(self):
        results_df = self.result_df
        if results_df is None:
            print("No results found. Run get_eps_mu_impedance_neff() first.")
            return  
        plt.plot(results_df["frequency"], results_df["impedance"], "go", label="$Z_{eff}$")
        plt.xlabel("Frequency")
        plt.legend()
        plt.grid(True)

    def plot_neff_vs_freqs(self):
        results_df = self.result_df
        if self.result_df is None:
            print("No results found. Run get_eps_mu_impedance_neff() first.")
            return
        plt.plot(results_df["frequency"], results_df["n_eff"], "yo", label="$n_{eff}$")
        plt.xlabel("Frequency")
        plt.legend()
        plt.grid(True)

def _retrieve_single_point(analyzer: FieldAnalyzer, k_idx, band, component_i, component_j, average_type, nonbloch, conversion_options,  overwrite):
    eps, mu, e_over_h_I, e_over_h_II = analyzer.get_eps_mu(k_idx, band, component_i, component_j, average_type, nonbloch, conversion_options, overwrite=overwrite)
    impedance = analyzer.get_impedance(eps, mu)
    neff = analyzer.get_refractive_index(eps, mu)
    kmag, freq = analyzer.get_kmag_and_freq(k_idx, band)
    return {
        'k_index': k_idx,
        'band': band,
        'frequency': freq,
        'eps': eps,
        'mu': mu,
        'impedance': impedance,
        'n_eff': neff,
        'kmag': kmag,
        'e_over_h_I': e_over_h_I,
        'e_over_h_II': e_over_h_II,
    }
```

```python
import numpy as np
import pandas as pd
from mpl_toolkits.axes_grid1.inset_locator import inset_axes
from matplotlib.ticker import ScalarFormatter
import matplotlib.pyplot as plt

# ===========================================================================
# Helper functions for tick formatting and plotting
# ===========================================================================

def clean_format_axis(ax, axis='y', scilimits=(-2, -2)):
    """Automatically scales axis and places the multiplier in the offset text."""
    formatter = ScalarFormatter(useMathText=True)
    formatter.set_scientific(True)
    formatter.set_powerlimits(scilimits)
    
    if axis == 'y':
        ax.yaxis.set_major_formatter(formatter)
    else:
        ax.xaxis.set_major_formatter(formatter)

def extract_band_data_at_ky_zero(sim, band_num='band_6', ky_tolerance=1e-4):
    """Extract k-points and frequencies where ky≈0 (more robust)."""
    ky_zero_mask = [abs(kv[1]) < ky_tolerance for kv in sim['k_vecs']]
    kx_vals = np.array([sim['k_vecs'][i][0] for i, m in enumerate(ky_zero_mask) if m])
    freq = np.array([sim['bands'][band_num][i] for i, m in enumerate(ky_zero_mask) if m])
    
    # Sort by kx to ensure proper order
    sort_idx = np.argsort(kx_vals)
    return kx_vals[sort_idx], freq[sort_idx]

def plot_band_data(ax, sim, band_num='band_6', color='blue', marker='o', 
                   label='Band 6 Data', ms=6, subtract_dirac=False):
    """Plot band data points onto an axis."""
    kx_vals, freq = extract_band_data_at_ky_zero(sim, band_num)
    
    if len(kx_vals) == 0:
        print(f"WARNING: No data points found for band {band_num}!")
        return kx_vals, freq
    
    if subtract_dirac:
        gamma_idx = np.argmin(np.abs(kx_vals))
        f_shift = freq[gamma_idx]
    else:
        f_shift = 0
    
    ax.scatter(kx_vals, freq - f_shift, color=color, marker=marker, label=label, 
               s=ms**2, zorder=10, alpha=0.85, edgecolors='black', linewidth=0.5)
    return kx_vals, freq

def plot_polynomial_fit(ax, sim, kx_range=None, color='blue', label='Cubic fit', 
                       lw=1, subtract_dirac=False):
    """Plot polynomial fit for band 6."""
    poly = np.poly1d(sim['band6_poly_coeffs'])
    kx_vals, _ = extract_band_data_at_ky_zero(sim, 'band_6')
    
    if kx_range is None:
        kx_range = (kx_vals.min(), kx_vals.max())
    
    kx_fit = np.linspace(kx_range[0], kx_range[1], 100)
    
    if subtract_dirac:
        f_shift = poly(0)
    else:
        f_shift = 0
    
    ax.plot(kx_fit, poly(kx_fit) - f_shift, '-', label=label, lw=lw, color=color, 
            zorder=5, alpha=0.8)
    return kx_fit, poly

def plot_linear_approximation(ax, sim, kx_range=None, color='blue', 
                              label='Linear', lw=1, linestyle='--', alpha=0.6):
    """Plot linear tangent at Gamma."""
    poly = np.poly1d(sim['band6_poly_coeffs'])
    kx_vals, _ = extract_band_data_at_ky_zero(sim, 'band_6')
    
    if kx_range is None:
        kx_range = (kx_vals.min(), kx_vals.max())
    
    kx_fit = np.linspace(kx_range[0], kx_range[1], 100)
    first_deriv = poly.deriv()(0)
    
    ax.plot(kx_fit, first_deriv * kx_fit, linestyle, 
            label=f'$v_g={first_deriv:.3f}$', lw=lw, alpha=alpha, color=color, zorder=3)
    return first_deriv

# ===========================================================================
# Use selected_points from cell 34 with matching colors
# ===========================================================================




fig, ax = plt.subplots(figsize=ROW1_COL1_SIZE)

for idx, (sim, color, label) in enumerate(zip(selected_points, highlight_colors, point_labels)):
    if idx == len(selected_points) - 1:
        break
    plot_band_data(ax, sim, band_num='band_6', color=color, marker='.',
                   label='', ms=1, subtract_dirac=True)
    plot_polynomial_fit(ax, sim, color=color,
                        label=f'{label} ',
                        subtract_dirac=True)
    plot_linear_approximation(ax, sim, color=color, alpha=0.6,
                              label=None)
    
ax.set_xlabel(r'$k_x a/(2\pi)$', fontsize=12)
ax.set_ylabel(r'$(\omega - \omega_D) a/(2\pi c)$', fontsize=12, labelpad=10)
ax.set_ylim(0, 0.01)
ax.grid(True, linestyle='--', alpha=0.6)
clean_format_axis(ax, axis='y')

# Inset with zoom
ax_inset = inset_axes(ax,
                      width="35%",
                      height="80%",
                      loc='upper left',
                      bbox_to_anchor=(0.1, 0.4, 0.85, 0.5),
                      bbox_transform=ax.transAxes)
kx_zoom_range = (-0.001, 0.01)
for idx, (sim, color) in enumerate(zip(selected_points, highlight_colors)):
    if idx == len(selected_points) - 1:
        break
    plot_band_data(ax_inset, sim, band_num='band_6', color=color, marker='.', ms=1,
                   subtract_dirac=True)
    plot_polynomial_fit(ax_inset, sim, kx_range=kx_zoom_range, color=color,
                        subtract_dirac=True)
    plot_linear_approximation(ax_inset, sim, kx_range=kx_zoom_range, color=color)


ax_inset.set_xlim(kx_zoom_range[0], kx_zoom_range[1])
ax_inset.set_ylim(0, 0.0015)
ax_inset.grid(True, linestyle='--', alpha=0.5)
ax_inset.tick_params(axis='both', which='major', labelsize=8)
clean_format_axis(ax_inset, axis='y')

ax.legend(loc='upper center', bbox_to_anchor=(0.5, -0.4), ncol=4, fontsize=7, frameon=True)
fig.suptitle("Band 6 Dispersion — Selected Points", y=0.98, fontsize = 10)
plt.subplots_adjust(left=0.15, right=0.95, top=0.85, bottom=0.30)

plt.savefig(os.path.join(figure_path, "band6_dispersion_selected_points.png"))
plt.show()
```

```python
import phc_nzi.field_analyzer as fa

# ===========================================================================
# Effective Parameter Extraction  (ε_eff, μ_eff, n_eff, Z_eff)
# Uses the selected_points from the previous cell (Left 50%, Peak, Right 50%, Baseline Tail)
# ===========================================================================

# --- Build df_selected from selected_points (determined in previous cell) ---
df_selected = pd.DataFrame([
    {'r1': p['r1'], 'r2': p['r2'], 'filling_factor': p['ff']}
    for p in selected_points
])

print(f"Using {len(selected_points)} points from previous cell:")
for i, p in enumerate(selected_points):
    print(f"  [{i}] {point_labels[i]:<15} | r1={p['r1']:.4f}, r2={p['r2']:.4f}, ff={p['ff']:.4f}")

# --- Configuration for nonbloch fields ---
k_points_eff = {
    "k_points_values": [
        mp.Vector3(0.002, 0, 0),
        mp.Vector3(0.2, 0, 0),
    ],
    "k_points_label": [
        r"$\Gamma$",
        r"$k_x$",
    ],
}

config_eff = dict(
    resolution=64,
    num_bands=6,
    k_points=k_points_eff["k_points_values"],
    k_points_interpolation_factor=20,
    extra_runner_command=(
        # "fix-efield-phase "
        "output-hfield output-nonbloch-efield-y output-nonbloch-hfield-z"
    ),
)
mpb_config_eff = MPBSchemeConfigurator(photonic_crystal_2D, ["te"], **config_eff)
script_eff = mpb_config_eff.get_scheme_config(join_newline=True)



# --- Cache file for effective parameters ---
eff_cache_file = os.path.join(data_rooth, f"{name_2D_optimized}_effective_params_selected.csv")

# --- Build one df_eff per point directly ---
df_eff_per_point = {}

if os.path.exists(eff_cache_file) and not SIMULATE:
    # ---- Load cached results and split immediately ----
    _df_all = pd.read_csv(eff_cache_file)
    for col_name in ['eps', 'mu', 'impedance', 'n_eff']:
        _df_all[col_name] = _df_all[f'{col_name}_re'] + 1j * _df_all[f'{col_name}_im']
    print(f"Loaded cached effective parameters from:\n  {eff_cache_file}")

    for i, (p, lbl) in enumerate(zip(selected_points, point_labels)):
        mask = (np.isclose(_df_all['r1'], p['r1'], atol=1e-6) &
                np.isclose(_df_all['r2'], p['r2'], atol=1e-6))
        df_eff_per_point[lbl] = _df_all[mask].copy().reset_index(drop=True)
    del _df_all
else:
    # ---- Compute from field data, one df per point ----
    save_parts = []
    for i, row in df_selected.iterrows():
        r1_pt = row['r1']
        r2_pt = row['r2']
        lbl = point_labels[i]
        tag = f"r1_{r1_pt:.4f}_r2_{r2_pt:.4f}".replace(".", "p")
        sim_name = f"{name_2D_optimized}_eff_{tag}"
        sim_eff = Simulation(
            simulation_name=sim_name,
            script=script_eff,
            directory=os.path.join(data_rooth, sim_name),
        )

        if SIMULATE:
            print(f"[{i+1}/{len(df_selected)}] Running: r1={r1_pt:.4f}, r2={r2_pt:.4f}")
            sim_eff.run_hpc(
                mpb_command_line_params=dict(r1=r1_pt, r2=r2_pt), mpi=True
            )
        else:
            print(f"[{i+1}/{len(df_selected)}] Using existing data "
                  f"(r1={r1_pt:.4f}, r2={r2_pt:.4f})")

        try:
            analyzer_eff = FieldAnalyzer(sim_eff, [4, 6], "te", "x")
            data_eff = analyzer_eff.get_eps_mu_impedance_neff("y", "z", plot=True, average_type="2D")
            plt.title(f"Effective Parameters - {lbl}")
            data_eff['r1'] = r1_pt
            data_eff['r2'] = r2_pt
            data_eff['filling_factor'] = row['filling_factor']
            df_eff_per_point[lbl] = data_eff.reset_index(drop=True)
            save_parts.append(data_eff)
            plt.figure()
            plt.title(f"Effective Parameters - {lbl}")
            analyzer_eff.plot_eps_vs_freqs()
            analyzer_eff.plot_mu_vs_freqs()
            plt.show()
            plt.figure()
            plt.title(f"Effective Parameters - {lbl}")
            analyzer_eff.plot_neff_vs_freqs()

            plt.figure()
            plt.title(f"Effective Parameters - {lbl}")
            analyzer_eff.plot_impedance_vs_freqs()

        except Exception as e:
            print(f"  ⚠ Error at r1={r1_pt:.4f}, r2={r2_pt:.4f}: {e}")

    print(f"\nExtracted effective parameters for {len(df_eff_per_point)}/{len(df_selected)} points")

    # ---- Save to CSV ----
    df_save = pd.concat(save_parts, ignore_index=True)
    for col_name in ['eps', 'mu', 'impedance', 'n_eff']:
        df_save[f'{col_name}_re'] = np.real(df_save[col_name])
        df_save[f'{col_name}_im'] = np.imag(df_save[col_name])
    df_save = df_save.drop(columns=['eps', 'mu', 'impedance', 'n_eff'])
    df_save.to_csv(eff_cache_file, index=False)
    print(f"Saved effective parameters to:\n  {eff_cache_file}")

# --- Compute shifted frequency per point ---
for lbl, df_pt in df_eff_per_point.items():
    k1_b4 = df_pt.loc[(df_pt['band'] == 4) & (df_pt['kmag'] == df_pt['kmag'].min()), 'frequency'].values[0]
    k1_b6 = df_pt.loc[(df_pt['band'] == 6) & (df_pt['kmag'] == df_pt['kmag'].min()), 'frequency'].values[0]
    omega_D = 0.5*(k1_b4 + k1_b6)
    df_pt['omega_dirac'] = omega_D
    df_pt['freq_shifted'] = df_pt['frequency'] - omega_D
    print(f"  {lbl}: {len(df_pt)} rows, ω_D = {omega_D:.6f}")

# --- Reconstruct merged df_eff for downstream compatibility ---
df_eff = pd.concat(df_eff_per_point.values(), ignore_index=True)
print(f"\ndf_eff_per_point: {len(df_eff_per_point)} points | df_eff: {len(df_eff)} total rows")

```

```python
## Fix phase jumps

def unwrap_region_swaps(df):
    """
    Detects pi phase jumps that swap Region I and Region II in the MPB data
    and restores continuity by swapping them back.
    """
    df = df.sort_values('freq_shifted').reset_index(drop=True)
    
    # Ensure e_over_h columns are complex (they may be strings after CSV round-trip)
    for col in ['e_over_h_I', 'e_over_h_II']:
        if df[col].dtype == object:
            df[col] = df[col].apply(lambda x: complex(x.replace(' ', '')) if isinstance(x, str) else complex(x))
    
    e_h_1 = df['e_over_h_I'].values.copy().astype(complex)
    e_h_2 = df['e_over_h_II'].values.copy().astype(complex)
    
    # ...existing code...
    corr_1 = np.zeros_like(e_h_1, dtype=complex)
    corr_2 = np.zeros_like(e_h_2, dtype=complex)
    
    corr_1[0] = e_h_1[0]
    corr_2[0] = e_h_2[0]
    
    for i in range(1, len(df)):
        dist_no_swap = np.abs(e_h_1[i] - corr_1[i-1]) + np.abs(e_h_2[i] - corr_2[i-1])
        dist_swap = np.abs(e_h_2[i] - corr_1[i-1]) + np.abs(e_h_1[i] - corr_2[i-1])
        
        if dist_swap < dist_no_swap:
            corr_1[i] = e_h_2[i]
            corr_2[i] = e_h_1[i]
        else:
            corr_1[i] = e_h_1[i]
            corr_2[i] = e_h_2[i]
            
    df['e_over_h_I_corr'] = corr_1
    df['e_over_h_II_corr'] = corr_2
    return df

def recompute_effective_parameters(df):
    """
    Recomputes eps, mu, impedance, and n_eff using the strictly real 
    part of the corrected E/H ratios to avoid branch cut artifacts.
    """
    # Force purely real values to eliminate branch cut noise
    e_h_1 = np.real(df['e_over_h_I_corr'].values)
    e_h_2 = np.real(df['e_over_h_II_corr'].values)
    
    k_mag_over_freq = df['kmag'].values / df['frequency'].values
    
    mu_I = e_h_1
    mu_II = e_h_2
    eps_I = 1.0 / e_h_1
    eps_II = 1.0 / e_h_2
    
    # Compute global mu and eps
    mu_corr = np.emath.sqrt(mu_I) * np.emath.sqrt(mu_II) * k_mag_over_freq
    eps_corr = np.emath.sqrt(eps_I) * np.emath.sqrt(eps_II) * k_mag_over_freq
    
    impedance_corr = np.emath.sqrt(mu_corr / eps_corr)
    n_eff_corr = np.emath.sqrt(eps_corr) * np.emath.sqrt(mu_corr)
    
    df['eps_corr'] = eps_corr
    df['mu_corr'] = mu_corr
    df['impedance_corr'] = impedance_corr
    df['n_eff_corr'] = n_eff_corr
    
    # Calculate log(1/cost) FOM based on impedance mismatch
    cost = np.abs(df['impedance_corr'] - 1.0) + 1e-12
    df['fom_log'] = np.log(1.0 / cost)
    
    return df

def process_all_points(df_dictionary):
    """Applies unwrapping and recomputation cleanly to all (r1, r2) points."""
    processed_dfs = {}
    for point, df in df_dictionary.copy().items():
        df = unwrap_region_swaps(df)
        df = recompute_effective_parameters(df)
        processed_dfs[point] = df
    return processed_dfs    

# Process the entire dataframe at once
df_eff_corrected_per_point = process_all_points(df_eff_per_point)

```

```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import concurrent.futures

# ===========================================================================
# 1. Symmetry Projection Functions
# ===========================================================================
def unmix_Hz_field(Hz_mixed):
    """Separates pseudo-scalar Hz into 1D (Monopole) and 2D (Dipole/E) components."""
    Hz_flip = np.rot90(Hz_mixed, k=2, axes=(0, 1))
    Hz_1D = 0.5 * (Hz_mixed + Hz_flip)
    Hz_E  = 0.5 * (Hz_mixed - Hz_flip)
    return Hz_1D, Hz_E

# ===========================================================================
# 2. Symmetry-Aware Parameter Extraction
# ===========================================================================
def extract_symmetry_aware_parameters(sim, k_idx, freq_dict, kmag, opt):
    """
    Dynamically tracks the Monopole (DNG) and Dipole (DPS) modes using 
    symmetry projection, bypassing MPB's frequency-based band swapping.
    """
    bands_to_check = [4, 5, 6]
    
    # Storage for projection weights
    w_1D_list = []
    w_E_list = []
    fields = {}

    # 1. Load and project fields for all candidate bands
    for b in bands_to_check:
        Hz = sim.load_and_convert_field_data(k_idx, b, "z", "te", "h", nonbloch=True, conversion_options=opt)
        Ey = sim.load_and_convert_field_data(k_idx, b, "y", "te", "e", nonbloch=True, conversion_options=opt)
        
        if Hz.ndim == 3: Hz = Hz[:, :, Hz.shape[2]//2]
        if Ey.ndim == 3: Ey = Ey[:, :, Ey.shape[2]//2]
            
        Hz_1D, Hz_E = unmix_Hz_field(Hz)
        
        # Calculate projection energy (weights)
        w_1D = np.sum(np.abs(Hz_1D)**2)
        w_E = np.sum(np.abs(Hz_E)**2)
        
        w_1D_list.append(w_1D)
        w_E_list.append(w_E)
        fields[b] = {'Hz': Hz, 'Ey': Ey}

    # 2. Identify the true physical modes
    band_mono = bands_to_check[np.argmax(w_1D_list)]
    band_dipo = bands_to_check[np.argmax(w_E_list)]
    
    results = []
    
    # 3. Helper to compute E/H averages and effective params using the Re() fix
    def compute_params(b_target, sign_mask):
        Hz_tgt = fields[b_target]['Hz']
        Ey_tgt = fields[b_target]['Ey']
        freq_tgt = freq_dict[b_target]
        
        # Dynamic Masking based on real Hz
        mask_I = Hz_tgt.real >= 0
        mask_II = Hz_tgt.real < 0
        
        # Regional Averages
        E_avg_I = np.mean(Ey_tgt[mask_I]) if np.any(mask_I) else 0
        H_avg_I = np.mean(Hz_tgt[mask_I]) if np.any(mask_I) else 1e-12
        E_avg_II = np.mean(Ey_tgt[mask_II]) if np.any(mask_II) else 0
        H_avg_II = np.mean(Hz_tgt[mask_II]) if np.any(mask_II) else 1e-12
        
        # Get purely real E/H to avoid branch cut artifacts
        e_h_1 = np.real(E_avg_I / H_avg_I)
        e_h_2 = np.real(E_avg_II / H_avg_II)
        
        k_over_f = kmag / freq_tgt if freq_tgt > 0 else 0
        
        # Compute params
        mu_corr = np.emath.sqrt(e_h_1) * np.emath.sqrt(e_h_2) * k_over_f
        eps_corr = np.emath.sqrt(1.0/e_h_1) * np.emath.sqrt(1.0/e_h_2) * k_over_f
        
        # Enforce physical sign based on the mode
        mu_true = np.abs(mu_corr) * sign_mask
        eps_true = np.abs(eps_corr) * sign_mask
        z_true = np.sqrt(mu_true / eps_true)
        n_true = np.sqrt(eps_true * mu_true) * sign_mask
        
        # FOM requirement: log(1/cost)
        cost = np.abs(z_true - 1.0) + 1e-12
        fom_log = np.log(1.0 / cost)
        
        return {
            'band_tracked': b_target,
            'frequency': freq_tgt,
            'kmag': kmag,
            'eps': eps_true,
            'mu': mu_true,
            'impedance': z_true,
            'n_eff': n_true,
            'fom_log': fom_log
        }

    # Compute for Monopole (DNG regime -> negative)
    res_mono = compute_params(band_mono, sign_mask=-1)
    res_mono['mode_type'] = 'Monopole'
    results.append(res_mono)
    
    # Compute for Dipole (DPS regime -> positive)
    res_dipo = compute_params(band_dipo, sign_mask=1)
    res_dipo['mode_type'] = 'Dipole'
    results.append(res_dipo)
    
    return results
```

```python
# ===========================================================================
# Phase diagnostic: check angle(H_z / E_y) across k-points
# A ~π jump between consecutive k-points means the phase fixing is
# inconsistent (fix-hfield-phase + fix-efield-phase pick signs independently)
# ===========================================================================
import phc_nzi.field_analyzer as fa
import numpy as np
import matplotlib.pyplot as plt

# Use the peak-point simulation (already created in the previous cell)
analyzer_diag = fa.FieldAnalyzer(sim_peak, [4, 6], "te", "x")
k_indices = analyzer_diag.get_k_indices()

fig, axes = plt.subplots(2, 2, figsize=(14, 8), constrained_layout=True)

for b_i, band in enumerate([4, 6]):
    phases = []
    ratios_re = []
    ratios_im = []

    for k_idx in k_indices:
        ey = analyzer_diag.load_efield_data(k_idx, band, "y", nonbloch=True)
        hz = analyzer_diag.load_hfield_data(k_idx, band, "z", nonbloch=True)

        # Spatial average (unit cell average)
        ey_avg = ey.mean()
        hz_avg = hz.mean()

        ratio = hz_avg / ey_avg
        phases.append(np.angle(ratio))
        ratios_re.append(ratio.real)
        ratios_im.append(ratio.imag)

    phases = np.array(phases)
    ratios_re = np.array(ratios_re)
    ratios_im = np.array(ratios_im)

    # Detect jumps
    dphase = np.diff(phases)
    dphase_wrapped = (dphase + np.pi) % (2 * np.pi) - np.pi
    jumps = np.abs(dphase_wrapped) > np.pi / 2
    n_jumps = jumps.sum()

    # --- Left column: phase vs k-index ---
    ax = axes[b_i, 0]
    ax.plot(k_indices, np.degrees(phases), 'o-', ms=4, lw=1.2, label=f'Band {band}')
    if n_jumps > 0:
        jump_k = np.array(k_indices)[1:][jumps]
        jump_ph = np.degrees(phases[1:][jumps])
        ax.scatter(jump_k, jump_ph, s=120, facecolors='none', edgecolors='red',
                   linewidths=2, zorder=5, label=f'{n_jumps} jump(s) > 90°')
    ax.set_xlabel('k-index')
    ax.set_ylabel(r'$\arg(\langle H_z \rangle / \langle E_y \rangle)$ [deg]')
    ax.set_title(f'Band {band}: phase of $H_z/E_y$  —  {"⚠ JUMPS" if n_jumps else "✓ smooth"}')
    ax.axhline(0, color='gray', ls='--', lw=0.6)
    ax.grid(True, alpha=0.3)
    ax.legend(fontsize=8)

    # --- Right column: Re and Im of ratio ---
    ax2 = axes[b_i, 1]
    ax2.plot(k_indices, ratios_re, 'o-', ms=4, lw=1.2, label=r'Re$(H_z/E_y)$')
    ax2.plot(k_indices, ratios_im, 's-', ms=4, lw=1.2, label=r'Im$(H_z/E_y)$')
    ax2.set_xlabel('k-index')
    ax2.set_ylabel(r'$\langle H_z \rangle / \langle E_y \rangle$')
    ax2.set_title(f'Band {band}: real & imag parts')
    ax2.axhline(0, color='gray', ls='--', lw=0.6)
    ax2.grid(True, alpha=0.3)
    ax2.legend(fontsize=8)

    print(f"Band {band}: {n_jumps} phase jump(s) > 90° out of {len(k_indices)-1} steps")
    if n_jumps:
        for jk in np.array(k_indices)[1:][jumps]:
            print(f"  → jump at k_index={jk} (Δφ = {np.degrees(dphase_wrapped[jk - k_indices[0] - 1]):.1f}°)")

fig.suptitle(f'Phase consistency diagnostic — r1={peak_point["r1"]:.4f}, r2={peak_point["r2"]:.4f}',
             fontsize=13, fontweight='bold')
plt.show()
```

```python
# ===========================================================================
# Compute ε_eff and μ_eff FROM SCRATCH via simple field averaging
# (no masking, no sign splitting — just spatial averages)
#
#   Z_eff = <E_y> / <H_z>     (impedance from field ratio)
#   n_eff = k / ω              (refractive index from dispersion)
#   ε_eff = n_eff / Z_eff
#   μ_eff = n_eff * Z_eff
# ===========================================================================
import numpy as np
import matplotlib.pyplot as plt
import phc_nzi.field_analyzer as fa

bands = [4, 6]
analyzer_scratch = fa.FieldAnalyzer(sim_peak, bands, "te", "x")
k_indices = list(analyzer_scratch.get_k_indices())
df_freq = analyzer_scratch.df  # frequency DataFrame

results_scratch = []

for band in bands:
    for k_idx in k_indices:
        # Load nonbloch fields
        ey = analyzer_scratch.load_efield_data(k_idx, band, "y", nonbloch=True)
        hz = analyzer_scratch.load_hfield_data(k_idx, band, "z", nonbloch=True)

        # Simple spatial average over the unit cell
        ey_avg = ey.mean()
        hz_avg = hz.mean()

        # Impedance
        Z = ey_avg / hz_avg

        # k and ω from the band structure
        kmag, freq = analyzer_scratch.get_kmag_and_freq(k_idx, band)

        # Refractive index from dispersion
        n = kmag / freq if freq != 0 else np.nan

        # Effective parameters
        eps_eff = n / Z if Z != 0 else np.nan
        mu_eff  = n * Z

        results_scratch.append({
            'k_index': k_idx, 'band': band,
            'kmag': kmag, 'freq': freq,
            'ey_avg': ey_avg, 'hz_avg': hz_avg,
            'Z': Z, 'n': n,
            'eps': eps_eff, 'mu': mu_eff,
        })

import pandas as pd
df_scratch = pd.DataFrame(results_scratch)

# Compute Dirac frequency (average of bands 4 & 6 at smallest k)
k_min = df_scratch['k_index'].min()
f4_0 = df_scratch.loc[(df_scratch['band'] == 4) & (df_scratch['k_index'] == k_min), 'freq'].values[0]
f6_0 = df_scratch.loc[(df_scratch['band'] == 6) & (df_scratch['k_index'] == k_min), 'freq'].values[0]
omega_D_scratch = 0.5 * (f4_0 + f6_0)
df_scratch['freq_shifted'] = df_scratch['freq'] - omega_D_scratch

print(f"ω_D = {omega_D_scratch:.6f}")
print(f"Total data points: {len(df_scratch)}")

# ===========================================================================
# Plot: compare old (FieldAnalyzer masking) vs new (simple average)
# ===========================================================================
fig, axes = plt.subplots(2, 3, figsize=(18, 10), constrained_layout=True)

band_colors = {4: 'C0', 6: 'C1'}

for band in bands:
    c = band_colors[band]

    # --- old data from df_eff (masking method) ---
    old = sub_peak[sub_peak['band'] == band].sort_values('freq_shifted')
    # --- new data from simple averaging ---
    new = df_scratch[df_scratch['band'] == band].sort_values('freq_shifted')

    # Row 0: ε_eff
    axes[0, 0].plot(old['freq_shifted'], np.real(old['eps']),
                    'o-', color=c, ms=4, lw=1, alpha=0.5, label=f'Band {band} (old/masked)')
    axes[0, 0].plot(new['freq_shifted'], np.real(new['eps']),
                    's--', color=c, ms=4, lw=1.5, label=f'Band {band} (new/avg)')

    # Row 0: μ_eff
    axes[0, 1].plot(old['freq_shifted'], np.real(old['mu']),
                    'o-', color=c, ms=4, lw=1, alpha=0.5, label=f'Band {band} (old/masked)')
    axes[0, 1].plot(new['freq_shifted'], np.real(new['mu']),
                    's--', color=c, ms=4, lw=1.5, label=f'Band {band} (new/avg)')

    # Row 0: Z_eff
    axes[0, 2].plot(old['freq_shifted'], np.real(old['impedance']),
                    'o-', color=c, ms=4, lw=1, alpha=0.5, label=f'Band {band} (old/masked)')
    axes[0, 2].plot(new['freq_shifted'], np.real(new['Z']),
                    's--', color=c, ms=4, lw=1.5, label=f'Band {band} (new/avg)')

    # Row 1: imaginary parts
    axes[1, 0].plot(new['freq_shifted'], np.imag(new['eps']),
                    's--', color=c, ms=4, lw=1.5, label=f'Band {band} Im(ε)')
    axes[1, 1].plot(new['freq_shifted'], np.imag(new['mu']),
                    's--', color=c, ms=4, lw=1.5, label=f'Band {band} Im(μ)')
    axes[1, 2].plot(new['freq_shifted'], np.imag(new['Z']),
                    's--', color=c, ms=4, lw=1.5, label=f'Band {band} Im(Z)')

titles = [r'Re($\varepsilon_{\mathrm{eff}}$)', r'Re($\mu_{\mathrm{eff}}$)', r'Re($Z_{\mathrm{eff}}$)',
          r'Im($\varepsilon_{\mathrm{eff}}$)', r'Im($\mu_{\mathrm{eff}}$)', r'Im($Z_{\mathrm{eff}}$)']
for i, ax in enumerate(axes.flat):
    ax.set_xlabel(r'$(\omega - \omega_D)\, a/2\pi c$')
    ax.set_title(titles[i])
    ax.axhline(0, color='gray', ls='--', lw=0.6)
    ax.axvline(0, color='gray', ls=':', lw=0.5, alpha=0.5)
    ax.grid(True, alpha=0.3)
    ax.legend(fontsize=7, loc='best')

fig.suptitle(f'Effective parameters from scratch (simple averaging) vs old (masked)\n'
             f'r1={peak_point["r1"]:.4f}, r2={peak_point["r2"]:.4f}',
             fontsize=13, fontweight='bold')
plt.show()

# Print a summary table
print("\n--- Simple-average results (band 4, first 5 k-points) ---")
print(df_scratch[df_scratch['band'] == 4][['k_index', 'kmag', 'freq', 'freq_shifted', 'Z', 'n', 'eps', 'mu']].head().to_string())
```

```python
import phc_nzi.field_analyzer as fa

# ===========================================================================
# Effective Parameter Extraction  (ε_eff, μ_eff, n_eff, Z_eff)
# Uses the selected_points from the previous cell (Left 50%, Peak, Right 50%, Baseline Tail)
# ===========================================================================

# --- Build df_selected from selected_points (determined in previous cell) ---
df_selected = pd.DataFrame([
    {'r1': p['r1'], 'r2': p['r2'], 'filling_factor': p['ff']}
    for p in selected_points
])

print(f"Using {len(selected_points)} points from previous cell:")
for i, p in enumerate(selected_points):
    print(f"  [{i}] {labels[i]:<15} | r1={p['r1']:.4f}, r2={p['r2']:.4f}, ff={p['ff']:.4f}")

# --- Configuration for nonbloch fields ---
k_points_eff = {
    "k_points_values": [
        mp.Vector3(0, 0, 0),
        mp.Vector3(0.05, 0, 0),
    ],
    "k_points_label": [
        r"$\Gamma$",
        r"$k_x$",
    ],
}

config_eff = dict(
    resolution=RESOLUTION,
    num_bands=6,
    k_points=k_points_eff["k_points_values"],
    k_points_interpolation_factor=40,
    extra_runner_command=(
        "fix-hfield-phase "
        "output-hfield output-nonbloch-efield-y output-nonbloch-hfield-z"
    ),
)
mpb_config_eff = MPBSchemeConfigurator(photonic_crystal_2D, ["te"], **config_eff)
script_eff = mpb_config_eff.get_scheme_config(join_newline=True)

SIMULATE_EFF = True   # ← Set True to run simulations on HPC

# --- Cache file for effective parameters ---
eff_cache_file = os.path.join(data_rooth, f"{name_2D_optimized}_effective_params_selected.csv")

# --- Build one df_eff per point directly ---
df_eff_per_point = {}

if os.path.exists(eff_cache_file) and not SIMULATE_EFF:
    # ---- Load cached results and split immediately ----
    _df_all = pd.read_csv(eff_cache_file)
    for col_name in ['eps', 'mu', 'impedance', 'n_eff']:
        _df_all[col_name] = _df_all[f'{col_name}_re'] + 1j * _df_all[f'{col_name}_im']
    print(f"Loaded cached effective parameters from:\n  {eff_cache_file}")

    for i, (p, lbl) in enumerate(zip(selected_points, labels)):
        mask = (np.isclose(_df_all['r1'], p['r1'], atol=1e-6) &
                np.isclose(_df_all['r2'], p['r2'], atol=1e-6))
        df_eff_per_point[lbl] = _df_all[mask].copy().reset_index(drop=True)
    del _df_all
else:
    # ---- Compute from field data, one df per point ----
    save_parts = []
    for i, row in df_selected.iterrows():
        r1_pt = row['r1']
        r2_pt = row['r2']
        lbl = labels[i]
        tag = f"r1_{r1_pt:.4f}_r2_{r2_pt:.4f}".replace(".", "p")
        sim_name = f"{name_2D_optimized}_eff_{tag}"
        sim_eff = Simulation(
            simulation_name=sim_name,
            script=script_eff,
            directory=os.path.join(data_rooth, sim_name),
        )

        if SIMULATE_EFF:
            print(f"[{i+1}/{len(df_selected)}] Running: r1={r1_pt:.4f}, r2={r2_pt:.4f}")
            sim_eff.run_hpc(
                mpb_command_line_params=dict(r1=r1_pt, r2=r2_pt), mpi=True
            )
        else:
            print(f"[{i+1}/{len(df_selected)}] Using existing data "
                  f"(r1={r1_pt:.4f}, r2={r2_pt:.4f})")

        try:
            analyzer_eff = FieldAnalyzer(sim_eff, [4, 5, 6], "te", "x")
            data_eff = analyzer_eff.get_eps_mu_impedance_neff("y", "z", plot=False)
            data_eff['r1'] = r1_pt
            data_eff['r2'] = r2_pt
            data_eff['filling_factor'] = row['filling_factor']
            df_eff_per_point[lbl] = data_eff.reset_index(drop=True)
            save_parts.append(data_eff)
        except Exception as e:
            print(f"  ⚠ Error at r1={r1_pt:.4f}, r2={r2_pt:.4f}: {e}")

    print(f"\nExtracted effective parameters for {len(df_eff_per_point)}/{len(df_selected)} points")

    # ---- Save to CSV ----
    df_save = pd.concat(save_parts, ignore_index=True)
    for col_name in ['eps', 'mu', 'impedance', 'n_eff']:
        df_save[f'{col_name}_re'] = np.real(df_save[col_name])
        df_save[f'{col_name}_im'] = np.imag(df_save[col_name])
    df_save = df_save.drop(columns=['eps', 'mu', 'impedance', 'n_eff'])
    df_save.to_csv(eff_cache_file, index=False)
    print(f"Saved effective parameters to:\n  {eff_cache_file}")

# --- Compute shifted frequency per point ---
for lbl, df_pt in df_eff_per_point.items():
    k1_b4 = df_pt.loc[(df_pt['band'] == 4) & (df_pt['k_index'] == df_pt['k_index'].min()), 'frequency'].values[0]
    k1_b6 = df_pt.loc[(df_pt['band'] == 6) & (df_pt['k_index'] == df_pt['k_index'].min()), 'frequency'].values[0]
    omega_D = 0.5 * (k1_b4 + k1_b6)
    df_pt['omega_dirac'] = omega_D
    df_pt['freq_shifted'] = df_pt['frequency'] - omega_D
    print(f"  {lbl}: {len(df_pt)} rows, ω_D = {omega_D:.6f}")

# --- Reconstruct merged df_eff for downstream compatibility ---
df_eff = pd.concat(df_eff_per_point.values(), ignore_index=True)
print(f"\ndf_eff_per_point: {len(df_eff_per_point)} points | df_eff: {len(df_eff)} total rows")
```

```python
import concurrent.futures
import time

# Assume 'name' and 'script_slab' are defined elsewhere
pop = numproc // 2
h_array = [0.25, 0.3, 0.35, 0.4,0.45, 0.5, 0.55]


df_array= []
def run_optimization(h):
    opt_name = simulation_slab_opt.simulation_name + f"_p64_h_{h:0.3f}"
    optimizer = MPIdeOptimizator(
        simulation_name=opt_name,
        scheme_script=simulation_slab_opt.script,
        maxiter=15,
        param_names=["r1", "r2"],
        param_bounds=[(0.15, 0.35), (0.1, 0.35)],
        polarization="zeven",
        de_options={"strategy": "rand1bin", "popsize": pop},
        bands=[8, 9, 10],
        height_slab=h,
        directory=os.path.join(data_rooth, opt_name)
    )
    # Submit the LSF job.
    try:
        #optimizer.erease_data_file()
        pass
    except:
        pass
    #optimizer.submit_lsf_job(lsf_config)
    try:
        data_file = optimizer.data_file
        analyzer = OptimizationDataAnalyzer(data_file)
        # plt.figure()
        # analyzer.plot_raw_data(plot_inverse_cost=True, points_only=True, use_logscale=True)        
        # plt.close()
        # plt.figure()
        # analyzer.plot_optimization_points_freq_dirac(points_only=True)
        
        # plt.close()

        df = analyzer.get_points_above_treshold(1200)
        df["h"] = h
        df_array.append(df)

    except Exception as e:
        print(f"An error occurred while analyzing the data: {e}")

#Use ThreadPoolExecutor to run jobs concurrently with a small delay between job submissions.
with concurrent.futures.ThreadPoolExecutor() as executor:
    futures = []
    for h in h_array:
        futures.append(executor.submit(run_optimization, h))
        time.sleep(1)  # Add a small delay between submissions
    concurrent.futures.wait(futures)
```
